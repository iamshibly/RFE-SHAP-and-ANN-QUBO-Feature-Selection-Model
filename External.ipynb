```python
"""
Leakage-safe ovarian cancer ML pipeline
======================================

This script is designed to replace the two separate notebooks with one end-to-end,
leakage-controlled pipeline.

What it does
------------
1. Loads the user-uploaded Journal of Cancer supplementary dataset: jcav16p3192s1.xlsx.
2. Renames its clinical columns into the canonical names used by this pipeline.
3. Splits raw data into 80 percent training and 20 percent held-out test BEFORE
   imputation, binning, statistical filtering, feature selection, tuning, or fitting.
4. Fits preprocessing only on the active training data:
   - robust object biomarker parsing, including comma strings and censored values
     like >5000 or <0.5;
   - mean imputation learned from training only;
   - original numeric harmonization by multiplying float variables by 100 and
     converting to integer-compatible values;
   - original variable-specific discretization logic: log, quantile, equal-width,
     binary-preserved.
5. Runs leakage-safe nested CV:
   outer training fold -> fold-fitted preprocessing -> fold-specific Kruskal-Wallis
   -> fold-specific Cramer's V redundancy removal -> fold-specific feature selection
   -> inner hyperparameter tuning -> outer validation evaluation.
6. Runs both feature-selection branches:
   - Stat-RS: RFE plus SHAP using strict intersection only.
   - Stat-AQ: ANN permutation saliency plus quantum-inspired QUBO subset optimization,
     using strict union only. QUBO coefficients, weights, sampler,
     reads, seed, and objective are saved.
7. Evaluates DT, LR, SVM, KNN, LGBM, and XGB for both feature-selection branches.
8. Reports nested-CV and held-out-test metrics: sensitivity, specificity, PPV, NPV,
   F1, AUC, log loss, Brier score, confusion matrix counts, and bootstrap CIs.
   The QUBO branch is strict: OpenJij SQASampler is required and there is no
   classical fallback, so saved Stat-AQ outputs prove quantum-inspired SQA was used.
9. Creates plots:
   - Cramer's V heatmap;
   - selected-feature correlation heatmap;
   - ANN permutation strip and radar plots;
   - QUBO selected-feature ridge plot;
   - nested-CV and held-out-test ROC curves;
   - confusion matrices;
   - held-out-test calibration curve for each selector selected base/best model only;
   - selection stability plots across outer folds.
10. Adds a sensitivity-analysis section comparing the main discretized representation
    against a continuous standardized representation and spline-expanded LR.

Run in Colab or locally
-----------------------
    pip install openpyxl scipy scikit-learn seaborn matplotlib pandas numpy joblib shap xgboost lightgbm tensorflow openjij dimod
    python leakage_safe_ovarian_pipeline.py

Upload jcav16p3192s1.xlsx into the Colab Files panel. The expected path is:
    DATA_FILE = "/content/jcav16p3192s1.xlsx"

Important scientific note
-------------------------
This supplementary spreadsheet contains 391 patient-level rows: 189 epithelial ovarian
cancer cases and 202 benign ovarian tumor controls. The spreadsheet itself contains nine
usable predictors (Age, BMI, Menopause, CA125, HE4, NLR, PLR, MLR, SII) plus patient ID
and the binary group label. Stage and histology are described in the paper but are not
included as patient-level columns in this spreadsheet.
"""

from __future__ import annotations

import os
import re
import json
import math
import time
import random
import warnings
import subprocess
import sys
import shutil

# Global scientific aliases. These are also imported inside some functions,
# but keeping them global prevents notebook-cell execution NameError issues.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from dataclasses import dataclass, asdict
from pathlib import Path
from typing import Any, Dict, Iterable, List, Optional, Sequence, Tuple

warnings.filterwarnings("ignore")

INSTALL_MISSING_PACKAGES = True
RANDOM_STATE = 42
DATA_FILE: Optional[str] = "/content/jcav16p3192s1.xlsx"
OUTPUT_DIR = "ovarian_jcav391_leakage_safe_outputs"
CLEAN_OUTPUT_DIR = True
TARGET_COL = "TYPE"
TEST_SIZE = 0.20
OUTER_SPLITS = 10
INNER_SPLITS = 5
N_ITER_SEARCH = 30
BOOTSTRAP_N = 2000
BOOTSTRAP_SEED = 42
BOOTSTRAP_METRIC_CI_LEVEL = 0.95
SAVE_FIGS = True
SHOW_FIGS = False
PRINT_ALL_OUTPUTS = True
DISPLAY_ALL_SAVED_IMAGES_AT_END = False
DISPLAY_PER_FIGURE_AS_CREATED = True
PLOT_CALIBRATION_FOR_EACH_MODEL = False
RUN_LR_BASELINES_FOR_EACH_SELECTOR = False  # LR baselines run only when LR is the selected base model.
CREATE_OUTER_FOLD_DIAGNOSTIC_PLOTS = False
CREATE_FINAL_SELECTOR_DIAGNOSTIC_PLOTS = True
CALIBRATION_N_BINS = 7
CALIBRATION_BIN_STRATEGY = "quantile"



# Same fundamental process from the original notebooks/manuscript.
DROP_COLS_PRE_SPECIFIED = ["SUBJECT_ID"]
BINARY_COLS = ["TYPE", "Menopause"]
LOG_BIN_COLS = [
    "AFP", "ALP", "ALT", "AST", "CA125", "CA19-9", "CEA",
    "DBIL", "IBIL", "TBIL", "CREA", "GGT", "UA", "HE4",
]
QUANTILE_BIN_COLS = [
    "EO#", "EO%", "MONO#", "MONO%", "PCT", "PDW", "PLT", "BASO#", "BASO%",
]
EQUAL_BIN_COLS = [
    "AG", "Age", "ALB", "BUN", "Ca", "CL", "CO2CP", "GLO", "GLU.", "HCT", "HGB",
    "K", "LYM#", "LYM%", "MCH", "MCV", "Mg", "MPV", "Na", "PHOS", "RBC", "RDW", "TP",
]
CLINICALLY_PROTECTED_FEATURES = ["CA125", "HE4", "Menopause", "AFP", "ALB", "DBIL"]

# Feature-selection controls.
KRUSKAL_ALPHA = 0.05
CRAMERS_V_THRESHOLD = 0.90
RFE_TOP_N = 20
SHAP_TOP_N = 20
STAT_RS_COMBINE = "intersection"  # strict: RFE-SHAP uses intersection only. No union fallback.
STAT_RS_MIN_FEATURES = 1
ANN_EPOCHS = 10
ANN_BATCH_SIZE = 32
ANN_POSITIVE_IMPORTANCE_ONLY = True
ANN_MIN_FEATURES = 5
QUBO_ALPHA = 1.0
QUBO_BETA = 0.5
QUBO_LAMBDA = 0.2
QUBO_CARDINALITY_GAMMA = 0.0  # Set >0 and QUBO_TARGET_K if you want a hard-ish panel size.
QUBO_TARGET_K: Optional[int] = None
QUBO_NUM_READS = 100
# Reviewer-safe setting: Stat-AQ must use OpenJij simulated quantum annealing.
# If OpenJij is unavailable or fails, the script stops instead of silently falling back
# to a classical simulated annealer.
REQUIRE_OPENJIJ_SQA = True
STAT_AQ_COMBINE = "union"  # strict: ANN-QUBO uses union only. No intersection/fallback rule.
STAT_AQ_MIN_FEATURES = 1

# Optional clinically meaningful cutoffs for sensitivity analysis.
# Fill with validated thresholds only if supplied by a clinician/source.
# Example format: CLINICAL_CUTOFFS = {"CA125": [35], "HE4": [70, 140]}
CLINICAL_CUTOFFS: Dict[str, List[float]] = {}

RUN_STAT_AQ_XGB_VARIANT_BASELINES = True
RUN_CLINICAL_BIOMARKER_BASELINES = True
XAI_FOCUS_SELECTOR = "Stat-AQ"
XAI_MAX_DISPLAY_FEATURES = 20
XAI_BACKGROUND_SIZE = 100
XAI_EVAL_SIZE = 70


def ensure_packages() -> None:
    """Install optional packages in notebook/Colab style environments if missing."""
    if not INSTALL_MISSING_PACKAGES:
        return
    required = {
        "openpyxl": "openpyxl",
        "scipy": "scipy",
        "sklearn": "scikit-learn",
        "seaborn": "seaborn",
        "matplotlib": "matplotlib",
        "pandas": "pandas",
        "numpy": "numpy",
        "joblib": "joblib",
        "shap": "shap",
        "xgboost": "xgboost",
        "lightgbm": "lightgbm",
        "tensorflow": "tensorflow",
        "openjij": "openjij",
        "dimod": "dimod",
        "lime": "lime",
    }
    missing = []
    for import_name, package_name in required.items():
        try:
            __import__(import_name)
        except Exception:
            missing.append(package_name)
    if missing:
        print("Installing missing packages:", missing)
        subprocess.check_call([sys.executable, "-m", "pip", "install", "--quiet"] + missing)


def import_runtime_packages() -> Dict[str, Any]:
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns
    from scipy.stats import kruskal, chi2_contingency
    from sklearn.base import BaseEstimator, ClassifierMixin, clone
    from sklearn.calibration import calibration_curve
    from sklearn.compose import ColumnTransformer
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.feature_selection import RFE
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import LogisticRegression
    from sklearn.metrics import (
        accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
        log_loss, roc_curve, confusion_matrix, ConfusionMatrixDisplay,
        classification_report, brier_score_loss,
    )
    from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler, LabelEncoder, SplineTransformer
    from sklearn.svm import SVC
    from sklearn.tree import DecisionTreeClassifier
    import joblib

    # Optional imports.
    try:
        import shap
    except Exception:
        shap = None
    try:
        import tensorflow as tf
    except Exception:
        tf = None
    try:
        import xgboost as xgb
        from xgboost import XGBClassifier
    except Exception:
        xgb = None
        XGBClassifier = None
    try:
        from lightgbm import LGBMClassifier
    except Exception:
        LGBMClassifier = None
    try:
        import openjij as oj
    except Exception:
        oj = None
    try:
        import dimod
    except Exception:
        dimod = None

    return locals()


def set_seed(seed: int = RANDOM_STATE) -> None:
    random.seed(seed)
    try:
        import numpy as np
        np.random.seed(seed)
    except Exception:
        pass
    try:
        import tensorflow as tf
        tf.random.set_seed(seed)
    except Exception:
        pass


def make_output_dirs(base: str) -> Dict[str, Path]:
    root = Path(base)
    if CLEAN_OUTPUT_DIR and root.exists():
        shutil.rmtree(root)
    dirs = {
        "root": root,
        "figures": root / "figures",
        "tables": root / "tables",
        "models": root / "models",
        "features": root / "features",
        "logs": root / "logs",
    }
    for p in dirs.values():
        p.mkdir(parents=True, exist_ok=True)
    return dirs


def _display_image(path: Path) -> None:
    try:
        from IPython.display import display, Image
        display(Image(filename=str(path)))
    except Exception:
        print(f"[image saved] {path}")


def _display_html(path: Path) -> None:
    try:
        from IPython.display import display, HTML
        display(HTML(path.read_text(encoding="utf-8")))
    except Exception:
        print(f"[html saved] {path}")


def save_or_show(fig, path: Path) -> None:
    if SAVE_FIGS:
        fig.savefig(path, dpi=300, bbox_inches="tight")
        print(f"[saved figure] {path}")
        if DISPLAY_PER_FIGURE_AS_CREATED:
            _display_image(path)
    if SHOW_FIGS:
        fig.show()
    else:
        import matplotlib.pyplot as plt
        plt.close(fig)




def should_make_selector_diagnostic_plot(label: str) -> bool:
    text = str(label)
    if "outer_fold" in text:
        return bool(CREATE_OUTER_FOLD_DIAGNOSTIC_PLOTS)
    return bool(CREATE_FINAL_SELECTOR_DIAGNOSTIC_PLOTS)


def wilson_ci(k: int, n: int, z: float = 1.959963984540054) -> Tuple[float, float]:
    if n <= 0:
        return (np.nan, np.nan)
    phat = k / n
    denom = 1 + z * z / n
    center = (phat + z * z / (2 * n)) / denom
    margin = z * math.sqrt((phat * (1 - phat) + z * z / (4 * n)) / n) / denom
    return max(0.0, center - margin), min(1.0, center + margin)

def print_block(title: str) -> None:
    print("\n" + "=" * 100)
    print(title)
    print("=" * 100)


def print_df(title: str, df: Any, max_rows: Optional[int] = None) -> None:
    if not PRINT_ALL_OUTPUTS:
        return
    print_block(title)
    if df is None:
        print("None")
        return
    try:
        show_df = df if max_rows is None else df.head(max_rows)
        with pd.option_context("display.max_rows", None if max_rows is None else max_rows,
                               "display.max_columns", None, "display.width", 240,
                               "display.max_colwidth", 160):
            print(show_df.to_string(index=False))
    except Exception:
        print(df)


def metric_line(prefix: str, row: Dict[str, Any]) -> None:
    if not PRINT_ALL_OUTPUTS:
        return
    keys = ["Accuracy", "Sensitivity_Recall", "Specificity", "PPV_Precision", "NPV", "F1", "ROC_AUC", "Log_Loss", "Brier", "TN", "FP", "FN", "TP"]
    parts = []
    for k in keys:
        if k in row:
            v = row[k]
            try:
                if isinstance(v, (int, np.integer)):
                    parts.append(f"{k}={int(v)}")
                else:
                    parts.append(f"{k}={float(v):.4f}")
            except Exception:
                parts.append(f"{k}={v}")
    print(prefix + " | " + ", ".join(parts))


def print_selector_details(selector_name: str, selector_info: Dict[str, Any]) -> None:
    if not PRINT_ALL_OUTPUTS:
        return
    print_block(f"{selector_name} FEATURE-SELECTION OUTPUT")
    for key in ["rfe_features", "shap_features", "ann_features", "qubo_features", "selected_features", "combine_used"]:
        if key in selector_info:
            print(f"{key}: {selector_info[key]}")
    if "qubo_config" in selector_info:
        qc = selector_info["qubo_config"]
        keep = ["objective", "alpha_relevance_reward", "beta_redundancy_penalty", "lambda_sparsity_penalty", "num_reads", "random_state", "sampler", "best_energy", "required_solver", "no_classical_fallback"]
        print("qubo_config:")
        for k in keep:
            if k in qc:
                print(f"  {k}: {qc[k]}")


def print_preprocessing_details(
    label: str,
    prep: "LeakSafePreprocessor",
    X_train_raw: Any,
    X_train_prep: Any,
    kruskal_features: List[str],
    cv_dropped: List[str],
    X_filtered: Any,
    selected_features: List[str],
) -> None:
    if not PRINT_ALL_OUTPUTS:
        return
    print_block(f"{label}: PREPROCESSING AND FILTERING STEPS FITTED ON TRAINING DATA ONLY")
    print("Order used:")
    print("  1. Raw data was already split into 80% training and 20% held-out test before this step.")
    print("  2. Drop predefined non-carried variables inside active training data only.")
    print("  3. Parse object biomarkers inside active training data only.")
    print("  4. Learn mean-imputation values from active training data only.")
    print("  5. Apply numeric harmonization: float/object-derived variables x100 then integer-compatible conversion.")
    print("  6. Learn discretization/binning edges from active training data only.")
    print("  7. Transform validation/test using the fitted training-only preprocessing object.")
    print("  8. Kruskal-Wallis on active training data only.")
    print("  9. Cramer's V redundancy filtering on active training data only.")
    print(" 10. Feature-selection branch on active training data only.")
    print(f"Raw active-training shape before preprocessing: {X_train_raw.shape}")
    print(f"Preprocessed active-training shape after dropping/imputation/harmonization/binning: {X_train_prep.shape}")
    drop_existing = [c for c in DROP_COLS_PRE_SPECIFIED if c in X_train_raw.columns]
    print(f"Dropped predefined columns: {drop_existing}")
    print(f"Representation: {prep.representation}")
    print(f"Scale x100 columns ({len(prep.scale100_cols_)}): {prep.scale100_cols_}")
    audit_df = prep.audit_dataframe()
    print_df(f"{label}: OBJECT/NUMERIC PARSING AUDIT", audit_df, max_rows=None)
    means_df = pd.DataFrame({"feature": list(prep.impute_means_.keys()), "train_only_imputation_mean": list(prep.impute_means_.values())})
    print_df(f"{label}: TRAINING-ONLY IMPUTATION VALUES", means_df, max_rows=None)
    bins_rows = []
    for col, edges in prep.bin_edges_.items():
        try:
            edge_list = [float(x) if np.isfinite(x) else str(x) for x in list(edges)]
        except Exception:
            edge_list = list(edges)
        if col in LOG_BIN_COLS:
            rule = "log-bin"
        elif col in QUANTILE_BIN_COLS:
            rule = "quantile-bin"
        elif col in EQUAL_BIN_COLS:
            rule = "equal-width-bin"
        elif col in BINARY_COLS:
            rule = "binary-preserved"
        else:
            rule = "default-equal-width-bin"
        bins_rows.append({"feature": col, "rule": rule, "train_only_edges": edge_list, "observed_encoded_categories": prep.feature_categories_.get(col, [])})
    print_df(f"{label}: TRAINING-ONLY BINNING/DISCRETIZATION DETAILS", pd.DataFrame(bins_rows), max_rows=None)
    print(f"Kruskal-Wallis retained features ({len(kruskal_features)}): {kruskal_features}")
    print(f"Cramer's V dropped redundant features ({len(cv_dropped)}): {cv_dropped}")
    print(f"Features after Cramer's V ({X_filtered.shape[1]}): {list(X_filtered.columns)}")
    print(f"Final selected features ({len(selected_features)}): {selected_features}")


def display_all_saved_images(out_dirs: Dict[str, Path]) -> None:
    print_block("ALL SAVED FIGURES, DISPLAYED SEPARATELY")
    fig_paths = sorted(out_dirs["figures"].glob("*.png"))
    if not fig_paths:
        print("No PNG figures found.")
        return
    for i, path in enumerate(fig_paths, start=1):
        print(f"\nFIGURE {i}/{len(fig_paths)}: {path.name}")
        _display_image(path)


def print_output_inventory(out_dirs: Dict[str, Path]) -> None:
    print_block("OUTPUT FILE INVENTORY")
    rows = []
    for sub in ["tables", "features", "figures", "models", "logs"]:
        for p in sorted(out_dirs[sub].glob("*")):
            rows.append({"folder": sub, "file": p.name, "path": str(p)})
    print_df("Saved outputs", pd.DataFrame(rows), max_rows=None)


def find_dataset_file(download_root: str) -> str:
    candidates = []
    for root, _, files in os.walk(download_root):
        for fn in files:
            lower = fn.lower()
            full = os.path.join(root, fn)
            if lower.endswith((".xlsx", ".xls", ".csv")):
                candidates.append(full)
    if not candidates:
        raise FileNotFoundError(f"No CSV/XLSX file found under {download_root}")

    preferred = []
    for p in candidates:
        name = os.path.basename(p).lower()
        if "supplementary" in name and "data" in name and ("1" in name or "one" in name):
            preferred.append(p)
    if preferred:
        return sorted(preferred)[0]
    return sorted(candidates)[0]


def load_raw_dataset() -> Tuple[Any, str]:
    import pandas as pd

    # In Google Colab, upload jcav16p3192s1.xlsx to the Files panel.
    # It will normally be available at /content/jcav16p3192s1.xlsx.
    data_path = DATA_FILE or "/content/jcav16p3192s1.xlsx"

    if not os.path.exists(data_path):
        # Small convenience fallback in case Colab renamed the file, e.g. jcav16p3192s1 (1).xlsx.
        content_dir = Path("/content")
        matches = sorted(content_dir.glob("jcav16p3192s1*.xlsx")) if content_dir.exists() else []
        if matches:
            data_path = str(matches[0])
        else:
            raise FileNotFoundError(
                "Dataset not found. Upload jcav16p3192s1.xlsx into the Colab Files panel. "
                "Expected path: /content/jcav16p3192s1.xlsx"
            )

    if data_path.lower().endswith((".xlsx", ".xls")):
        df = pd.read_excel(data_path, sheet_name="Sheet1")
    elif data_path.lower().endswith(".csv"):
        df = pd.read_csv(data_path)
    else:
        raise ValueError(f"Unsupported data file: {data_path}")

    # Strip whitespace around column names first.
    df.columns = [str(c).strip() for c in df.columns]

    # Canonical names used by the original pipeline.
    # Source coding:
    #   Groups: 1 = epithelial ovarian cancer (EOC), 0 = benign ovarian tumor.
    #   Menopausal state: 1 = yes, 2 = no.
    rename_map = {
        "Patients": "SUBJECT_ID",
        "Groups": "TYPE",
        "Age": "Age",
        "BMI": "BMI",
        "Menopausal state（1yse2no）": "Menopause",
        "CA125（U/mL）35": "CA125",
        "HE4（pmol/l）": "HE4",
        "NLR": "NLR",
        "PLR": "PLR",
        "MLR": "MLR",
        "SII": "SII",
    }
    df = df.rename(columns=rename_map)

    expected_cols = [
        "SUBJECT_ID", "TYPE", "Age", "BMI", "Menopause",
        "CA125", "HE4", "NLR", "PLR", "MLR", "SII",
    ]
    missing = [c for c in expected_cols if c not in df.columns]
    if missing:
        raise ValueError(
            "The uploaded spreadsheet does not match the expected jcav16p3192s1.xlsx schema. "
            f"Missing canonical columns: {missing}. Raw columns seen: {list(df.columns)}"
        )

    # Keep only the actual patient-level fields from this supplementary spreadsheet.
    df = df[expected_cols].copy()

    # Convert published menopause coding (1=yes, 2=no) to a true binary predictor (1=yes, 0=no).
    raw_menopause = pd.to_numeric(df["Menopause"], errors="coerce")
    unexpected_menopause = sorted(set(raw_menopause.dropna().unique()) - {1, 2})
    if unexpected_menopause:
        raise ValueError(f"Unexpected Menopause codes found: {unexpected_menopause}")
    df["Menopause"] = raw_menopause.map({1: 1, 2: 0})

    # Validate/normalize the target exactly as supplied by the spreadsheet.
    df["TYPE"] = pd.to_numeric(df["TYPE"], errors="raise").astype(int)
    unexpected_target = sorted(set(df["TYPE"].dropna().unique()) - {0, 1})
    if unexpected_target:
        raise ValueError(f"Unexpected target codes found in Groups/TYPE: {unexpected_target}")

    # Numeric coercion for the remaining predictors; the general parsing/imputation pipeline
    # still runs later inside each training fold exactly as before.
    for col in ["Age", "BMI", "CA125", "HE4", "NLR", "PLR", "MLR", "SII"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # Dataset-specific integrity check: the released supplement contains 391 patient rows.
    if len(df) != 391:
        print(f"WARNING: expected 391 patient rows, found {len(df)}.")

    print(
        "Loaded jcav16p3192s1.xlsx | "
        f"rows={len(df)} | EOC(TYPE=1)={(df['TYPE'] == 1).sum()} | "
        f"Benign(TYPE=0)={(df['TYPE'] == 0).sum()} | "
        f"usable predictors={len(expected_cols) - 2}"
    )

    return df, data_path


def standardize_target(y: Any) -> Any:
    import pandas as pd
    from sklearn.preprocessing import LabelEncoder

    y_series = pd.Series(y).copy()
    if y_series.dtype == "object" or str(y_series.dtype).startswith("category"):
        return pd.Series(LabelEncoder().fit_transform(y_series), index=y_series.index, name=TARGET_COL)
    return y_series.astype(int)


def print_and_save_dataset_report(df: Any, out_dir: Path, source_path: str) -> None:
    import pandas as pd

    report_path = out_dir / "dataset_report.txt"
    lines: List[str] = []
    lines.append("DATASET REPORT")
    lines.append("==============")
    lines.append(f"Source file: {source_path}")
    lines.append(f"Shape: {df.shape[0]} rows x {df.shape[1]} columns")
    lines.append("")
    lines.append("Columns:")
    lines.extend([f"  {i+1:02d}. {c} | dtype={df[c].dtype} | missing={df[c].isna().sum()}" for i, c in enumerate(df.columns)])
    lines.append("")
    if TARGET_COL in df.columns:
        vc = df[TARGET_COL].value_counts(dropna=False).sort_index()
        lines.append("Target distribution:")
        for k, v in vc.items():
            lines.append(f"  {k}: {v} ({v / len(df) * 100:.2f}%)")
    lines.append("")
    lines.append("Missing values per column:")
    lines.append(df.isna().sum().to_string())
    lines.append("")
    lines.append("Raw dtypes:")
    lines.append(df.dtypes.to_string())
    text = "\n".join(lines)
    print(text)
    report_path.write_text(text, encoding="utf-8")


@dataclass
class ParseAudit:
    column: str
    total_values: int
    missing_before: int
    parsed_missing_after: int
    comma_values: int
    quoted_or_tabbed_values: int
    greater_than_values: int
    less_than_values: int
    non_numeric_to_missing: int


def parse_numeric_with_audit(series: Any, column: str) -> Tuple[Any, ParseAudit]:
    """
    Robust numeric parsing for object-formatted biomarkers.

    Rules:
    - Strip whitespace, tabs, quotes, and commas.
    - Convert ">x" and ">=x" to x as a conservative lower-bound numeric value.
    - Convert "<x" and "<=x" to x as a conservative upper-bound numeric value.
    - Report how many values were affected.

    This avoids silently discarding biologically meaningful extreme strings.
    """
    import numpy as np
    import pandas as pd

    s = series.copy()
    missing_before = int(s.isna().sum())
    raw = s.astype("string")

    comma_values = int(raw.fillna("").str.contains(",", regex=False).sum())
    quoted_or_tabbed_values = int(raw.fillna("").str.contains(r"[\"'\t]", regex=True).sum())
    greater_than_values = int(raw.fillna("").str.contains(r"^\s*>\s*=?", regex=True).sum())
    less_than_values = int(raw.fillna("").str.contains(r"^\s*<\s*=?", regex=True).sum())

    cleaned = raw.str.strip()
    cleaned = cleaned.str.replace("\t", "", regex=False)
    cleaned = cleaned.str.replace("\"", "", regex=False)
    cleaned = cleaned.str.replace("'", "", regex=False)
    cleaned = cleaned.str.replace(",", "", regex=False)
    cleaned = cleaned.str.replace(r"^\s*[<>]=?\s*", "", regex=True)
    cleaned = cleaned.str.replace(r"[^0-9eE+\-.]", "", regex=True)
    parsed = pd.to_numeric(cleaned, errors="coerce")

    parsed_missing_after = int(parsed.isna().sum())
    non_numeric_to_missing = max(0, parsed_missing_after - missing_before)
    audit = ParseAudit(
        column=column,
        total_values=len(series),
        missing_before=missing_before,
        parsed_missing_after=parsed_missing_after,
        comma_values=comma_values,
        quoted_or_tabbed_values=quoted_or_tabbed_values,
        greater_than_values=greater_than_values,
        less_than_values=less_than_values,
        non_numeric_to_missing=non_numeric_to_missing,
    )
    return parsed, audit




def save_enhanced_dataset_diagnostics(df: Any, out_dirs: Dict[str, Path]) -> None:
    print_block("ENHANCED DATASET DIAGNOSTICS")
    diag_dir = out_dirs["tables"] / "dataset_diagnostics"
    diag_dir.mkdir(parents=True, exist_ok=True)

    n = len(df)
    missing_df = pd.DataFrame({
        "feature": df.columns,
        "dtype": [str(df[c].dtype) for c in df.columns],
        "missing_n": [int(df[c].isna().sum()) for c in df.columns],
        "missing_pct": [float(df[c].isna().mean() * 100.0) for c in df.columns],
        "unique_n": [int(df[c].nunique(dropna=True)) for c in df.columns],
    }).sort_values(["missing_pct", "feature"], ascending=[False, True])
    missing_df.to_csv(diag_dir / "missingness_dtype_unique_summary.csv", index=False)
    print_df("Missingness, dtype, and unique-value summary", missing_df, max_rows=None)

    object_rows = []
    for col in df.columns:
        if col == TARGET_COL:
            continue
        if str(df[col].dtype) == "object":
            _, audit = parse_numeric_with_audit(df[col], col)
            object_rows.append(asdict(audit))
    object_audit = pd.DataFrame(object_rows)
    if not object_audit.empty:
        object_audit.to_csv(diag_dir / "object_biomarker_parsing_audit_raw_dataset.csv", index=False)
        print_df("Object-formatted biomarker parsing audit on raw dataset", object_audit, max_rows=None)

    absent_clinical_metadata = [
        "stage", "Stage", "FIGO", "histology", "Histology", "grade", "Grade",
        "treatment", "Treatment", "benign", "Benign", "control_type", "inclusion",
        "exclusion", "source_population", "hospital", "scanner", "diagnosis_date",
        "collection_date", "timeframe", "year"
    ]
    present_meta = [c for c in absent_clinical_metadata if c in df.columns]
    missing_meta = [c for c in absent_clinical_metadata if c not in df.columns]
    metadata_df = pd.DataFrame({
        "metadata_item_checked": absent_clinical_metadata,
        "present_in_dataset": [c in present_meta for c in absent_clinical_metadata],
    })
    metadata_df.to_csv(diag_dir / "clinical_metadata_availability_check.csv", index=False)
    print_df("Clinical cohort metadata availability check", metadata_df, max_rows=None)
    if not present_meta:
        print("Clinical metadata note: stage, histology, treatment status, inclusion/exclusion criteria, source population, and dataset timeframe are not available as columns in this file.")

    duplicate_count = int(df.duplicated().sum())
    dup_df = pd.DataFrame([{"raw_row_duplicate_count": duplicate_count, "raw_rows": n}])
    dup_df.to_csv(diag_dir / "row_duplicate_summary.csv", index=False)
    print_df("Duplicate-row summary", dup_df, max_rows=None)

    if TARGET_COL in df.columns:
        y = standardize_target(df[TARGET_COL])
        target_df = y.value_counts().sort_index().rename_axis("class").reset_index(name="count")
        target_df["percent"] = target_df["count"] / len(y) * 100.0
        target_df.to_csv(diag_dir / "target_distribution.csv", index=False)
        print_df("Target distribution", target_df, max_rows=None)

        if "Menopause" in df.columns:
            tab = pd.crosstab(df["Menopause"], y, margins=True)
            tab.to_csv(diag_dir / "menopause_by_target_crosstab.csv")
            print_block("Menopause by target cross-tab")
            print(tab.to_string())

        key_cols = [c for c in ["CA125", "HE4", "AFP", "CEA", "ALB", "DBIL", "IBIL", "TBIL", "Age", "Menopause"] if c in df.columns]
        parsed = pd.DataFrame(index=df.index)
        for col in key_cols:
            parsed[col], _ = parse_numeric_with_audit(df[col], col)
        parsed[TARGET_COL] = y.values
        class_summary = parsed.groupby(TARGET_COL).agg(["count", "mean", "std", "median", "min", "max"]).T.reset_index()
        class_summary.columns = ["feature", "statistic"] + [f"class_{c}" for c in sorted(y.unique())]
        class_summary.to_csv(diag_dir / "key_biomarker_classwise_numeric_summary.csv", index=False)
        print_df("Key biomarker class-wise numeric summary", class_summary, max_rows=None)

def make_equal_edges(values: Any, bins: int) -> Any:
    import numpy as np
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    if arr.size == 0 or np.nanmin(arr) == np.nanmax(arr):
        return np.array([-np.inf, np.inf])
    edges = np.linspace(np.nanmin(arr), np.nanmax(arr), bins + 1)
    edges[0] = -np.inf
    edges[-1] = np.inf
    return np.unique(edges)


def make_quantile_edges(values: Any, q: int) -> Any:
    import numpy as np
    import pandas as pd
    arr = pd.Series(values).dropna().astype(float)
    if arr.empty or arr.nunique() <= 1:
        return np.array([-np.inf, np.inf])
    quantiles = np.linspace(0, 1, q + 1)
    edges = np.unique(np.nanquantile(arr, quantiles))
    if len(edges) <= 2:
        return np.array([-np.inf, np.inf])
    edges[0] = -np.inf
    edges[-1] = np.inf
    return edges


def apply_edges(values: Any, edges: Any) -> Any:
    import numpy as np
    import pandas as pd
    edges = np.asarray(edges, dtype=float)
    if len(edges) <= 2:
        return pd.Series(np.zeros(len(values), dtype=int), index=values.index)
    binned = pd.cut(values.astype(float), bins=edges, labels=False, include_lowest=True)
    binned = binned.fillna(0).astype(int)
    return binned


class LeakSafePreprocessor:
    """Fit-on-train-only preprocessing reproducing the original pipeline safely."""

    def __init__(self, representation: str = "binned"):
        if representation not in {"binned", "continuous", "clinical_cutoff"}:
            raise ValueError("representation must be binned, continuous, or clinical_cutoff")
        self.representation = representation
        self.columns_: List[str] = []
        self.scale100_cols_: List[str] = []
        self.impute_means_: Dict[str, float] = {}
        self.bin_edges_: Dict[str, Any] = {}
        self.parse_audits_: List[ParseAudit] = []
        self.feature_categories_: Dict[str, List[int]] = {}
        self.fitted_ = False

    def _clean_base(self, X: Any, fit: bool = False) -> Any:
        import numpy as np
        import pandas as pd

        Xc = X.copy()
        drop_existing = [c for c in DROP_COLS_PRE_SPECIFIED if c in Xc.columns]
        if drop_existing:
            Xc = Xc.drop(columns=drop_existing)

        # Keep only predictors. Do not allow target in preprocessor X.
        if TARGET_COL in Xc.columns:
            Xc = Xc.drop(columns=[TARGET_COL])

        if fit:
            self.columns_ = list(Xc.columns)
        else:
            missing_cols = [c for c in self.columns_ if c not in Xc.columns]
            if missing_cols:
                raise ValueError(f"Transform data is missing columns: {missing_cols}")
            Xc = Xc[self.columns_]

        parsed = pd.DataFrame(index=Xc.index)
        audits = []
        original_dtypes = Xc.dtypes.astype(str).to_dict()
        for col in Xc.columns:
            parsed_col, audit = parse_numeric_with_audit(Xc[col], col)
            parsed[col] = parsed_col
            audits.append(audit)

        if fit:
            self.parse_audits_ = audits
            # Original notebook converted object columns to float, then multiplied all float columns by 100.
            # Here, columns that were object or float before parsing are treated as scale100 columns.
            self.scale100_cols_ = [
                c for c in Xc.columns
                if ("float" in original_dtypes[c].lower()) or ("object" in original_dtypes[c].lower())
            ]
            self.impute_means_ = {}
            for col in parsed.columns:
                mean_val = parsed[col].mean()
                if not np.isfinite(mean_val):
                    mean_val = 0.0
                self.impute_means_[col] = float(mean_val)

        for col in parsed.columns:
            parsed[col] = parsed[col].fillna(self.impute_means_.get(col, 0.0))

        for col in self.scale100_cols_:
            if col in parsed.columns:
                parsed[col] = np.ceil(parsed[col].astype(float) * 100.0).astype(int)

        # Non-scaled values are still made integer-compatible when possible, preserving original process.
        for col in parsed.columns:
            if col not in self.scale100_cols_:
                parsed[col] = np.ceil(parsed[col].astype(float)).astype(int)

        return parsed

    def fit(self, X: Any, y: Optional[Any] = None) -> "LeakSafePreprocessor":
        import numpy as np
        import pandas as pd

        base = self._clean_base(X, fit=True)

        if self.representation == "binned":
            for col in base.columns:
                if col in BINARY_COLS:
                    continue
                if col in LOG_BIN_COLS:
                    vals = np.log1p(base[col].clip(lower=0).astype(float))
                    self.bin_edges_[col] = make_equal_edges(vals, bins=5)
                elif col in QUANTILE_BIN_COLS:
                    self.bin_edges_[col] = make_quantile_edges(base[col], q=4)
                elif col in EQUAL_BIN_COLS:
                    self.bin_edges_[col] = make_equal_edges(base[col], bins=5)
                else:
                    self.bin_edges_[col] = make_equal_edges(base[col], bins=5)
        elif self.representation == "clinical_cutoff":
            for col, cuts in CLINICAL_CUTOFFS.items():
                if col in base.columns:
                    edges = [-np.inf] + sorted([float(x) for x in cuts]) + [np.inf]
                    self.bin_edges_[col] = np.asarray(edges)

        transformed = self.transform(X)
        self.feature_categories_ = {
            col: sorted(pd.Series(transformed[col]).dropna().astype(int).unique().tolist())
            for col in transformed.columns
        }
        self.fitted_ = True
        return self

    def transform(self, X: Any) -> Any:
        import numpy as np
        import pandas as pd

        base = self._clean_base(X, fit=False)

        if self.representation == "continuous":
            return base.astype(float)

        out = pd.DataFrame(index=base.index)
        if self.representation == "binned":
            for col in base.columns:
                if col in BINARY_COLS:
                    out[col] = base[col].astype(int)
                elif col in LOG_BIN_COLS:
                    vals = np.log1p(base[col].clip(lower=0).astype(float))
                    out[col] = apply_edges(vals, self.bin_edges_[col])
                else:
                    out[col] = apply_edges(base[col], self.bin_edges_[col])
            return out.astype(int)

        if self.representation == "clinical_cutoff":
            for col in base.columns:
                if col in self.bin_edges_:
                    out[col] = apply_edges(base[col], self.bin_edges_[col])
                else:
                    out[col] = base[col].astype(float)
            return out

        raise ValueError(f"Unknown representation: {self.representation}")

    def fit_transform(self, X: Any, y: Optional[Any] = None) -> Any:
        self.fit(X, y)
        return self.transform(X)

    def audit_dataframe(self) -> Any:
        import pandas as pd
        return pd.DataFrame([asdict(a) for a in self.parse_audits_])


def split_raw_holdout(df: Any) -> Tuple[Any, Any, Any, Any]:
    from sklearn.model_selection import train_test_split

    if TARGET_COL not in df.columns:
        raise ValueError(f"Target column {TARGET_COL} not found. Columns: {list(df.columns)}")
    y = standardize_target(df[TARGET_COL])
    X = df.drop(columns=[TARGET_COL])
    X_train_raw, X_test_raw, y_train, y_test = train_test_split(
        X, y, test_size=TEST_SIZE, stratify=y, random_state=RANDOM_STATE
    )
    return X_train_raw, X_test_raw, y_train.reset_index(drop=True), y_test.reset_index(drop=True)


def reset_xy(X: Any, y: Any) -> Tuple[Any, Any]:
    return X.reset_index(drop=True), y.reset_index(drop=True)


def kruskal_filter(X: Any, y: Any, alpha: float = KRUSKAL_ALPHA) -> Tuple[List[str], Any]:
    import numpy as np
    import pandas as pd
    from scipy.stats import kruskal

    selected = []
    rows = []
    y_arr = pd.Series(y).reset_index(drop=True)
    Xr = X.reset_index(drop=True)
    for feature in Xr.columns:
        g0 = Xr.loc[y_arr == 0, feature]
        g1 = Xr.loc[y_arr == 1, feature]
        if g0.nunique(dropna=True) < 2 and g1.nunique(dropna=True) < 2:
            stat, p_value = np.nan, 1.0
        else:
            try:
                stat, p_value = kruskal(g0, g1)
            except Exception:
                stat, p_value = np.nan, 1.0
        keep = bool(p_value < alpha)
        rows.append({"feature": feature, "kruskal_H": stat, "p_value": p_value, "keep": keep})
        if keep:
            selected.append(feature)
    if not selected:
        # Safety fallback: choose top features by smallest p-value.
        tmp = pd.DataFrame(rows).sort_values("p_value")
        selected = tmp.head(min(10, len(tmp)))["feature"].tolist()
        tmp.loc[tmp["feature"].isin(selected), "keep"] = True
        rows = tmp.to_dict("records")
    return selected, pd.DataFrame(rows).sort_values("p_value")


def cramers_v(x: Any, y: Any) -> float:
    import numpy as np
    import pandas as pd
    from scipy.stats import chi2_contingency

    if pd.Series(x).nunique(dropna=True) <= 1 or pd.Series(y).nunique(dropna=True) <= 1:
        return 0.0
    table = pd.crosstab(x, y)
    if table.shape[0] <= 1 or table.shape[1] <= 1:
        return 0.0
    chi2 = chi2_contingency(table)[0]
    n = table.sum().sum()
    if n == 0:
        return 0.0
    phi2 = chi2 / n
    r, k = table.shape
    denom = min(k - 1, r - 1)
    if denom <= 0:
        return 0.0
    return float(np.sqrt(phi2 / denom))


def cramers_v_matrix(X: Any) -> Any:
    import pandas as pd
    cols = list(X.columns)
    mat = pd.DataFrame(index=cols, columns=cols, dtype=float)
    for c1 in cols:
        for c2 in cols:
            mat.loc[c1, c2] = cramers_v(X[c1], X[c2])
    return mat


def cramers_v_filter(
    X: Any,
    threshold: float = CRAMERS_V_THRESHOLD,
    protected: Sequence[str] = CLINICALLY_PROTECTED_FEATURES,
) -> Tuple[List[str], List[str], Any]:
    import numpy as np

    mat = cramers_v_matrix(X)
    binary_cols = [c for c in X.columns if set(X[c].dropna().unique()).issubset({0, 1})]
    upper = mat.where(np.triu(np.ones(mat.shape), k=1).astype(bool))

    dropped = []
    for col in upper.columns:
        if col in protected or col in binary_cols:
            continue
        high_corr = upper[col].dropna()
        if any(high_corr > threshold):
            dropped.append(col)
    kept = [c for c in X.columns if c not in dropped]
    return kept, dropped, mat


def plot_heatmap(matrix: Any, title: str, path: Path, figsize: Tuple[int, int] = (14, 10)) -> None:
    import matplotlib.pyplot as plt
    import seaborn as sns

    fig, ax = plt.subplots(figsize=figsize)
    sns.heatmap(matrix.astype(float), annot=True if matrix.shape[0] <= 25 else False,
                fmt=".2f", cmap="coolwarm", linewidths=0.5, cbar_kws={"shrink": 0.8}, ax=ax)
    ax.set_title(title, fontsize=16, fontweight="bold")
    ax.tick_params(axis="x", labelrotation=45)
    for label in ax.get_xticklabels() + ax.get_yticklabels():
        label.set_fontsize(10)
        label.set_fontweight("bold")
    fig.tight_layout()
    save_or_show(fig, path)


def plot_selected_correlation(X: Any, y: Any, title: str, path: Path) -> None:
    import pandas as pd
    df = X.copy()
    df[TARGET_COL] = list(y)
    corr = df.corr(numeric_only=True)
    plot_heatmap(corr, title, path, figsize=(18, 12))


def select_stat_rs(X: Any, y: Any, out_dirs: Dict[str, Path], label: str) -> Dict[str, Any]:
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.feature_selection import RFE

    runtime = import_runtime_packages()
    XGBClassifier = runtime["XGBClassifier"]
    shap = runtime["shap"]

    if X.shape[1] == 0:
        raise ValueError("Stat-RS received zero features")

    rfe_n = min(RFE_TOP_N, X.shape[1])
    rf = RandomForestClassifier(random_state=RANDOM_STATE, n_estimators=300)
    rfe = RFE(estimator=rf, n_features_to_select=rfe_n)
    rfe.fit(X, y)
    rfe_features = X.columns[rfe.support_].tolist()
    rfe_ranking = pd.DataFrame({
        "feature": X.columns,
        "rfe_rank": rfe.ranking_,
        "selected_by_rfe": rfe.support_.astype(int),
    }).sort_values(["selected_by_rfe", "rfe_rank"], ascending=[False, True])

    shap_n = min(SHAP_TOP_N, X.shape[1])
    if XGBClassifier is not None:
        xgb_model = XGBClassifier(eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1)
        xgb_model.fit(X, y)
        if shap is not None:
            try:
                explainer = shap.Explainer(xgb_model, X)
                shap_values = explainer(X)
                vals = shap_values.values
                if vals.ndim == 3:
                    vals = vals[:, :, 1]
                shap_importance = np.abs(vals).mean(axis=0)
            except Exception:
                shap_importance = xgb_model.feature_importances_
        else:
            shap_importance = xgb_model.feature_importances_
    else:
        rf2 = RandomForestClassifier(random_state=RANDOM_STATE, n_estimators=300)
        rf2.fit(X, y)
        shap_importance = rf2.feature_importances_

    shap_ranking = pd.DataFrame({
        "feature": X.columns,
        "shap_importance": shap_importance,
    }).sort_values("shap_importance", ascending=False)
    shap_features = shap_ranking.head(shap_n)["feature"].tolist()

    selected = sorted(set(rfe_features) & set(shap_features))
    combine_used = "strict_intersection"
    if len(selected) < STAT_RS_MIN_FEATURES:
        raise RuntimeError(
            f"Stat-RS strict intersection produced too few features in {label}: {selected}. "
            "This script does not use a union fallback because the paper method requires intersection only."
        )

    # Bubble plot similar to the original notebook.
    merged = shap_ranking.merge(rfe_ranking, on="feature", how="left")
    plot_df = merged[merged["feature"].isin(selected)].sort_values("shap_importance", ascending=True)
    if should_make_selector_diagnostic_plot(label) and not plot_df.empty:
        fig, ax = plt.subplots(figsize=(12, max(7, 0.4 * len(plot_df))))
        sizes = 400 + 5000 * (plot_df["shap_importance"] / (plot_df["shap_importance"].max() + 1e-12))
        ax.scatter(plot_df["shap_importance"], plot_df["feature"], s=sizes, alpha=0.75, edgecolors="black")
        ax.set_title(f"{label}: RFE-SHAP Feature Importance Bubble Plot", fontsize=16, fontweight="bold")
        ax.set_xlabel("Mean absolute SHAP / model importance", fontsize=12, fontweight="bold")
        ax.set_ylabel("Feature", fontsize=12, fontweight="bold")
        ax.grid(True, linestyle="--", alpha=0.5)
        fig.tight_layout()
        save_or_show(fig, out_dirs["figures"] / f"{label}_stat_rs_bubble.png")

    rfe_ranking.to_csv(out_dirs["features"] / f"{label}_rfe_ranking.csv", index=False)
    shap_ranking.to_csv(out_dirs["features"] / f"{label}_shap_ranking.csv", index=False)
    pd.DataFrame({"selected_feature": selected}).to_csv(out_dirs["features"] / f"{label}_stat_rs_selected.csv", index=False)

    return {
        "selector": "Stat-RS",
        "rfe_features": rfe_features,
        "shap_features": shap_features,
        "selected_features": selected,
        "combine_used": combine_used,
    }


class KerasWrapper:
    def __init__(self, model: Any):
        self.model = model
    def fit(self, X: Any, y: Any) -> "KerasWrapper":
        return self
    def predict(self, X: Any) -> Any:
        import numpy as np
        return (self.model.predict(X, verbose=0).reshape(-1) > 0.5).astype(int)
    def predict_proba(self, X: Any) -> Any:
        import numpy as np
        p = self.model.predict(X, verbose=0).reshape(-1, 1)
        return np.hstack([1 - p, p])


def ann_permutation_features(X: Any, y: Any, out_dirs: Dict[str, Path], label: str) -> Tuple[List[str], Any]:
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns
    from math import pi
    from sklearn.model_selection import train_test_split
    from sklearn.preprocessing import StandardScaler
    from sklearn.inspection import permutation_importance
    from sklearn.metrics import accuracy_score
    from sklearn.neural_network import MLPClassifier

    runtime = import_runtime_packages()
    tf = runtime["tf"]

    if X.shape[1] == 0:
        return [], pd.DataFrame(columns=["feature", "ann_importance"])

    X_train, X_val, y_train, y_val = train_test_split(
        X, y, stratify=y, test_size=0.20, random_state=RANDOM_STATE
    )
    scaler = StandardScaler()
    X_train_s = scaler.fit_transform(X_train)
    X_val_s = scaler.transform(X_val)

    if tf is not None:
        set_seed(RANDOM_STATE)
        model = tf.keras.Sequential([
            tf.keras.layers.Input(shape=(X.shape[1],)),
            tf.keras.layers.Dense(64, activation="relu"),
            tf.keras.layers.Dense(1, activation="sigmoid"),
        ])
        model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
        model.fit(X_train_s, y_train, epochs=ANN_EPOCHS, batch_size=ANN_BATCH_SIZE, verbose=0)
        wrapped = KerasWrapper(model)
        perm = permutation_importance(
            wrapped, X_val_s, y_val, n_repeats=10, random_state=RANDOM_STATE, scoring="accuracy"
        )
        importances = perm.importances_mean
    else:
        mlp = MLPClassifier(hidden_layer_sizes=(64,), max_iter=500, random_state=RANDOM_STATE)
        mlp.fit(X_train_s, y_train)
        perm = permutation_importance(
            mlp, X_val_s, y_val, n_repeats=10, random_state=RANDOM_STATE, scoring="accuracy"
        )
        importances = perm.importances_mean

    ranking = pd.DataFrame({"feature": X.columns, "ann_importance": importances}).sort_values(
        "ann_importance", ascending=False
    )
    if ANN_POSITIVE_IMPORTANCE_ONLY:
        selected = ranking.loc[ranking["ann_importance"] > 0, "feature"].tolist()
    else:
        selected = ranking.head(max(ANN_MIN_FEATURES, min(11, len(ranking))))["feature"].tolist()
    if len(selected) < ANN_MIN_FEATURES:
        selected = ranking.head(min(ANN_MIN_FEATURES, len(ranking)))["feature"].tolist()

    ranking.to_csv(out_dirs["features"] / f"{label}_ann_permutation_importance.csv", index=False)

    # Final-only strip and radar plots. Outer-fold plots are suppressed by default.
    if should_make_selector_diagnostic_plot(label):
        fig, ax = plt.subplots(figsize=(10, max(6, 0.35 * len(ranking))))
        sns.stripplot(x=ranking["ann_importance"].values, y=ranking["feature"].values, size=8, ax=ax)
        for i, val in enumerate(ranking["ann_importance"].values):
            ax.plot([0, val], [i, i], lw=2, alpha=0.5)
        ax.set_title(f"{label}: ANN Permutation Importance", fontsize=14, fontweight="bold")
        ax.set_xlabel("Importance score")
        ax.set_ylabel("Feature")
        fig.tight_layout()
        save_or_show(fig, out_dirs["figures"] / f"{label}_ann_permutation_strip.png")

    # Final-only radar plot, restricted to ANN-selected features first, then top ranked features if needed.
    radar_source = ranking[ranking["feature"].isin(selected)].copy()
    if radar_source.empty:
        radar_source = ranking.head(30).copy()
    radar_df = radar_source.head(30).copy()
    if False and should_make_selector_diagnostic_plot(label) and len(radar_df) >= 3:
        categories = radar_df["feature"].tolist()
        values = radar_df["ann_importance"].clip(lower=0).tolist()
        if max(values) <= 0:
            values = (radar_df["ann_importance"] - radar_df["ann_importance"].min() + 1e-6).tolist()
        n = len(categories)
        angles = [i / float(n) * 2 * pi for i in range(n)]
        values_closed = values + values[:1]
        angles_closed = angles + angles[:1]
        fig, ax = plt.subplots(figsize=(9, 9), subplot_kw={"polar": True})
        ax.plot(angles_closed, values_closed, linewidth=2, linestyle="solid")
        ax.fill(angles_closed, values_closed, alpha=0.25)
        ax.set_yticklabels([])
        ax.set_xticks(angles)
        ax.set_xticklabels(categories, fontsize=8)
        ax.set_title(f"{label}: Final ANN-Selected Feature Radar Plot", size=14, y=1.1, fontweight="bold")
        fig.tight_layout()
        save_or_show(fig, out_dirs["figures"] / f"{label}_ann_radar.png")

    return selected, ranking



def plot_stat_aq_final_selected_radar(ann_ranking: Any, selected_features: List[str], out_dirs: Dict[str, Path], label: str) -> None:
    import matplotlib.pyplot as plt
    from math import pi
    if not should_make_selector_diagnostic_plot(label):
        return
    if not selected_features or len(selected_features) < 3:
        return
    score_map = dict(zip(ann_ranking["feature"], ann_ranking["ann_importance"])) if ann_ranking is not None and not ann_ranking.empty else {}
    rows = []
    for feat in selected_features:
        rows.append({"feature": feat, "ann_importance": float(score_map.get(feat, 0.0))})
    df = pd.DataFrame(rows)
    df["plot_score"] = df["ann_importance"]
    min_score = float(df["plot_score"].min())
    if min_score < 0:
        df["plot_score"] = df["plot_score"] - min_score
    if float(df["plot_score"].max()) <= 0:
        df["plot_score"] = 1.0
    df = df.sort_values("plot_score", ascending=False).head(30)
    categories = df["feature"].tolist()
    values = df["plot_score"].tolist()
    n = len(categories)
    angles = [i / float(n) * 2 * pi for i in range(n)]
    fig, ax = plt.subplots(figsize=(9, 9), subplot_kw={"polar": True})
    ax.plot(angles + angles[:1], values + values[:1], linewidth=2, linestyle="solid")
    ax.fill(angles + angles[:1], values + values[:1], alpha=0.25)
    ax.set_yticklabels([])
    ax.set_xticks(angles)
    ax.set_xticklabels(categories, fontsize=8)
    ax.set_title(f"{label}: Final Stat-AQ Union-Selected Feature Radar Plot", size=14, y=1.1, fontweight="bold")
    fig.tight_layout()
    save_or_show(fig, out_dirs["figures"] / f"{label}_stat_aq_final_union_radar.png")
    df.to_csv(out_dirs["features"] / f"{label}_stat_aq_final_union_radar_scores.csv", index=False)


def plot_stat_rs_final_bubble_from_saved_rankings(out_dirs: Dict[str, Path], label: str, selected_features: List[str]) -> None:
    """Guaranteed final-only Stat-RS bubble plot for the final 80% training selected feature set."""
    import matplotlib.pyplot as plt
    import pandas as pd
    if not should_make_selector_diagnostic_plot(label):
        return
    shap_path = out_dirs["features"] / f"{label}_shap_ranking.csv"
    rfe_path = out_dirs["features"] / f"{label}_rfe_ranking.csv"
    if not shap_path.exists() or not rfe_path.exists() or not selected_features:
        return
    shap_ranking = pd.read_csv(shap_path)
    rfe_ranking = pd.read_csv(rfe_path)
    merged = shap_ranking.merge(rfe_ranking, on="feature", how="left")
    plot_df = merged[merged["feature"].isin(selected_features)].copy()
    if plot_df.empty:
        return
    plot_df = plot_df.sort_values("shap_importance", ascending=True)
    max_imp = float(plot_df["shap_importance"].max()) if float(plot_df["shap_importance"].max()) != 0 else 1.0
    sizes = 400 + 5000 * (plot_df["shap_importance"] / (max_imp + 1e-12))
    fig, ax = plt.subplots(figsize=(12, max(7, 0.45 * len(plot_df))))
    ax.scatter(plot_df["shap_importance"], plot_df["feature"], s=sizes, alpha=0.78, edgecolors="black")
    for _, row in plot_df.iterrows():
        ax.text(row["shap_importance"], row["feature"], f"  RFE rank={int(row.get('rfe_rank', 0))}", va="center", fontsize=8)
    ax.set_title(f"{label}: Final Stat-RS RFE-SHAP Intersection Feature Bubble Plot", fontsize=16, fontweight="bold")
    ax.set_xlabel("Mean absolute SHAP / model importance", fontsize=12, fontweight="bold")
    ax.set_ylabel("Selected feature", fontsize=12, fontweight="bold")
    ax.grid(True, linestyle="--", alpha=0.45)
    fig.tight_layout()
    save_or_show(fig, out_dirs["figures"] / f"{label}_FINAL_stat_rs_intersection_bubble.png")
    plot_df.to_csv(out_dirs["features"] / f"{label}_FINAL_stat_rs_intersection_bubble_values.csv", index=False)


def build_qubo(X: Any, y: Any) -> Tuple[Dict[Tuple[int, int], float], Any, Any, Dict[str, Any]]:
    import numpy as np
    import pandas as pd

    n = X.shape[1]
    y_arr = pd.Series(y).astype(float).values
    relevance = []
    for col in X.columns:
        x = X[col].astype(float).values
        if np.std(x) == 0 or np.std(y_arr) == 0:
            relevance.append(0.0)
        else:
            val = np.corrcoef(x, y_arr)[0, 1]
            if not np.isfinite(val):
                val = 0.0
            relevance.append(abs(float(val)))
    relevance = np.asarray(relevance)
    redundancy = np.abs(pd.DataFrame(X).corr().fillna(0.0).values)

    qubo: Dict[Tuple[int, int], float] = {}
    for i in range(n):
        qubo[(i, i)] = -QUBO_ALPHA * relevance[i] + QUBO_LAMBDA
    for i in range(n):
        for j in range(i + 1, n):
            qubo[(i, j)] = QUBO_BETA * redundancy[i, j]

    if QUBO_CARDINALITY_GAMMA > 0 and QUBO_TARGET_K is not None:
        # gamma * (sum_i z_i - k)^2 = gamma * [sum_i z_i + 2 sum_i<j z_i z_j - 2k sum_i z_i + k^2]
        for i in range(n):
            qubo[(i, i)] = qubo.get((i, i), 0.0) + QUBO_CARDINALITY_GAMMA * (1 - 2 * QUBO_TARGET_K)
        for i in range(n):
            for j in range(i + 1, n):
                qubo[(i, j)] = qubo.get((i, j), 0.0) + 2 * QUBO_CARDINALITY_GAMMA

    config = {
        "objective": "min E(z)=sum_i(-alpha*relevance_i+lambda)*z_i + sum_i<j beta*redundancy_ij*z_i*z_j + optional gamma*(sum z-k)^2",
        "alpha_relevance_reward": QUBO_ALPHA,
        "beta_redundancy_penalty": QUBO_BETA,
        "lambda_sparsity_penalty": QUBO_LAMBDA,
        "gamma_cardinality_penalty": QUBO_CARDINALITY_GAMMA,
        "target_k": QUBO_TARGET_K,
        "num_reads": QUBO_NUM_READS,
        "random_state": RANDOM_STATE,
        "required_solver": "openjij.SQASampler_simulated_quantum_annealing",
        "no_classical_fallback": bool(REQUIRE_OPENJIJ_SQA),
        "relevance_definition": "absolute Pearson/point-biserial correlation with target on fold-training data",
        "redundancy_definition": "absolute Pearson correlation between candidate predictors on fold-training data",
    }
    return qubo, relevance, redundancy, config


def solve_qubo(qubo: Dict[Tuple[int, int], float]) -> Tuple[Dict[int, int], str, float, Any]:
    """Solve the QUBO with OpenJij simulated quantum annealing only.

    This deliberately has no classical fallback. That makes the Stat-AQ branch
    reviewer-safe: if the output exists, it was produced by the declared
    quantum-inspired SQA solver, not by a silent fallback solver.
    """
    runtime = import_runtime_packages()
    oj = runtime["oj"]

    if oj is None:
        raise RuntimeError(
            "OpenJij is required for the quantum-inspired Stat-AQ branch. "
            "Install it with: pip install openjij. "
            "The script stops here to avoid silently replacing SQA with a classical fallback."
        )

    sampler = oj.SQASampler()
    try:
        sampleset = sampler.sample_qubo(Q=qubo, num_reads=QUBO_NUM_READS)
    except Exception as exc:
        raise RuntimeError(
            "OpenJij SQASampler failed. The script stops because REQUIRE_OPENJIJ_SQA=True "
            "and no classical fallback is allowed for reviewer-safe quantum-inspired reporting."
        ) from exc

    first = sampleset.first
    sample = {int(k): int(v) for k, v in first.sample.items()}
    energy = float(first.energy)
    return sample, "openjij.SQASampler_simulated_quantum_annealing", energy, sampleset


def qubo_features(X: Any, y: Any, out_dirs: Dict[str, Path], label: str) -> Tuple[List[str], Dict[str, Any]]:
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns

    qubo, relevance, redundancy, config = build_qubo(X, y)
    sample, sampler_name, energy, sampleset = solve_qubo(qubo)
    config["sampler"] = sampler_name
    config["best_energy"] = energy

    selected = [X.columns[i] for i in range(X.shape[1]) if int(sample.get(i, 0)) == 1]
    config["fallback_added_top_relevance"] = False
    config["strict_qubo_output_only"] = True

    pd.DataFrame({
        "feature": X.columns,
        "qubo_selected": [int(sample.get(i, 0)) for i in range(X.shape[1])],
        "relevance": relevance,
    }).to_csv(out_dirs["features"] / f"{label}_qubo_selection.csv", index=False)
    pd.DataFrame(redundancy, index=X.columns, columns=X.columns).to_csv(out_dirs["features"] / f"{label}_qubo_redundancy_matrix.csv")
    with open(out_dirs["features"] / f"{label}_qubo_config.json", "w", encoding="utf-8") as f:
        json.dump(config, f, indent=2)

    # Ridge/KDE plot of selected QUBO features.
    if should_make_selector_diagnostic_plot(label) and selected:
        fig, ax = plt.subplots(figsize=(12, max(7, 0.35 * len(selected))))
        for col in selected[:20]:
            try:
                sns.kdeplot(X[col].astype(float), fill=True, label=col, linewidth=1.5, ax=ax, alpha=0.35)
            except Exception:
                pass
        ax.set_title(f"{label}: Ridge/KDE Plot of QUBO Selected Features", fontsize=14, fontweight="bold")
        ax.set_ylabel("Density")
        ax.legend(loc="best", fontsize=8)
        fig.tight_layout()
        save_or_show(fig, out_dirs["figures"] / f"{label}_qubo_ridge.png")

    return selected, config


def select_stat_aq(X: Any, y: Any, out_dirs: Dict[str, Path], label: str) -> Dict[str, Any]:
    ann_features, ann_ranking = ann_permutation_features(X, y, out_dirs, label)
    q_features, q_config = qubo_features(X, y, out_dirs, label)
    selected = sorted(set(ann_features) | set(q_features))
    combine_used = "strict_union"
    if len(selected) < STAT_AQ_MIN_FEATURES:
        raise RuntimeError(
            f"Stat-AQ strict union produced too few features in {label}: {selected}. "
            "This script does not add fallback features because the paper method requires ANN-QUBO union only."
        )

    import pandas as pd
    pd.DataFrame({"selected_feature": selected}).to_csv(out_dirs["features"] / f"{label}_stat_aq_selected.csv", index=False)
    plot_stat_aq_final_selected_radar(ann_ranking, selected, out_dirs, label)
    return {
        "selector": "Stat-AQ",
        "ann_features": ann_features,
        "qubo_features": q_features,
        "selected_features": selected,
        "combine_used": combine_used,
        "qubo_config": q_config,
    }


def run_fold_preprocess_and_select(
    selector_name: str,
    X_train_fold_raw: Any,
    y_train_fold: Any,
    out_dirs: Dict[str, Path],
    label: str,
    representation: str = "binned",
) -> Tuple[LeakSafePreprocessor, List[str], Dict[str, Any], Any, Any, List[str], List[str], Any]:
    prep = LeakSafePreprocessor(representation=representation)
    X_train_prep = prep.fit_transform(X_train_fold_raw, y_train_fold)
    X_train_prep, y_train_fold = reset_xy(X_train_prep, y_train_fold)

    kruskal_features, kruskal_df = kruskal_filter(X_train_prep, y_train_fold)
    X_kw = X_train_prep[kruskal_features]
    kept, dropped, cv_mat = cramers_v_filter(X_kw)
    X_filtered = X_kw[kept]

    kruskal_df.to_csv(out_dirs["features"] / f"{label}_kruskal.csv", index=False)
    pd.DataFrame({"dropped_by_cramers_v": dropped}).to_csv(out_dirs["features"] / f"{label}_cramers_dropped.csv", index=False)
    cv_mat.to_csv(out_dirs["features"] / f"{label}_cramers_v_matrix.csv")

    if selector_name == "Stat-RS":
        selector_info = select_stat_rs(X_filtered, y_train_fold, out_dirs, label)
    elif selector_name == "Stat-AQ":
        selector_info = select_stat_aq(X_filtered, y_train_fold, out_dirs, label)
    else:
        raise ValueError("selector_name must be Stat-RS or Stat-AQ")

    selected_features = selector_info["selected_features"]
    if not selected_features:
        raise RuntimeError(f"{selector_name} produced no selected features in {label}")
    return prep, selected_features, selector_info, X_train_prep, X_filtered, kruskal_features, dropped, cv_mat


def model_spaces() -> Dict[str, Dict[str, Any]]:
    import numpy as np
    from sklearn.tree import DecisionTreeClassifier
    from sklearn.linear_model import LogisticRegression
    from sklearn.svm import SVC
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler

    runtime = import_runtime_packages()
    XGBClassifier = runtime["XGBClassifier"]
    LGBMClassifier = runtime["LGBMClassifier"]

    spaces: Dict[str, Dict[str, Any]] = {}

    spaces["DT"] = {
        "estimator": DecisionTreeClassifier(random_state=RANDOM_STATE),
        "params": {
            "criterion": ["gini", "entropy", "log_loss"],
            "splitter": ["best", "random"],
            "max_depth": [None, 5, 10, 15, 20, 25, 30],
            "min_samples_split": [2, 5, 10, 15],
            "min_samples_leaf": [1, 2, 4, 6],
            "max_features": [None, "sqrt", "log2"],
        },
    }

    spaces["LR"] = {
        "estimator": Pipeline([
            ("scaler", StandardScaler()),
            ("lr", LogisticRegression(random_state=RANDOM_STATE)),
        ]),
        "params": {
            "lr__penalty": ["l1", "l2", "elasticnet", None],
            "lr__C": np.logspace(-3, 3, 10),
            "lr__solver": ["lbfgs", "liblinear", "saga", "newton-cg"],
            "lr__max_iter": [100, 500, 1000],
            "lr__fit_intercept": [True, False],
            "lr__l1_ratio": [None, 0.25, 0.5, 0.75],
        },
    }

    spaces["SVM"] = {
        "estimator": Pipeline([
            ("scaler", StandardScaler()),
            ("svc", SVC(probability=True, random_state=RANDOM_STATE)),
        ]),
        "params": {
            "svc__C": np.logspace(-3, 3, 10),
            "svc__kernel": ["linear", "rbf", "poly", "sigmoid"],
            "svc__gamma": ["scale", "auto"] + list(np.logspace(-4, 1, 6)),
            "svc__degree": [2, 3, 4],
            "svc__coef0": [0.0, 0.1, 0.5, 1.0],
            "svc__probability": [True],
            "svc__shrinking": [True, False],
            "svc__tol": [1e-3, 1e-4],
            "svc__class_weight": [None, "balanced"],
        },
    }

    spaces["KNN"] = {
        "estimator": Pipeline([
            ("scaler", StandardScaler()),
            ("knn", KNeighborsClassifier()),
        ]),
        "params": {
            "knn__n_neighbors": list(range(3, 51, 2)),
            "knn__weights": ["uniform", "distance"],
            "knn__metric": ["euclidean", "manhattan", "minkowski"],
            "knn__algorithm": ["auto", "ball_tree", "kd_tree", "brute"],
            "knn__leaf_size": list(range(10, 51, 5)),
            "knn__p": [1, 2],
        },
    }

    if LGBMClassifier is not None:
        spaces["LGBM"] = {
            "estimator": LGBMClassifier(random_state=RANDOM_STATE, n_jobs=-1, verbose=-1),
            "params": {
                "n_estimators": [100, 200, 300, 400],
                "learning_rate": [0.01, 0.05, 0.1, 0.2],
                "max_depth": [-1, 3, 5, 7, 9],
                "num_leaves": [15, 31, 63, 127],
                "min_child_samples": [10, 20, 30, 50],
                "subsample": [0.6, 0.7, 0.8, 0.9, 1.0],
                "colsample_bytree": [0.6, 0.7, 0.8, 0.9, 1.0],
                "reg_alpha": [0.0, 0.1, 1.0, 5.0],
                "reg_lambda": [0.0, 0.1, 1.0, 5.0],
            },
        }
    else:
        print("LightGBM is unavailable. LGBM will be skipped.")

    if XGBClassifier is not None:
        spaces["XGB"] = {
            "estimator": XGBClassifier(eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1),
            "params": {
                "n_estimators": [100, 200, 300, 400, 500],
                "learning_rate": [0.01, 0.05, 0.1, 0.2],
                "max_depth": [3, 5, 7, 9, 11],
                "min_child_weight": [1, 3, 5, 7],
                "subsample": [0.6, 0.7, 0.8, 0.9, 1.0],
                "colsample_bytree": [0.6, 0.7, 0.8, 0.9, 1.0],
                "gamma": [0, 0.1, 0.25, 0.5],
                "reg_alpha": [0.0, 0.1, 1.0, 5.0],
                "reg_lambda": [0.0, 0.1, 1.0, 5.0],
            },
        }
    else:
        print("XGBoost is unavailable. XGB will be skipped.")

    return spaces


def tune_model(estimator: Any, params: Dict[str, Any], X: Any, y: Any) -> Any:
    from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV

    inner_cv = StratifiedKFold(n_splits=INNER_SPLITS, shuffle=True, random_state=RANDOM_STATE)
    search = RandomizedSearchCV(
        estimator=estimator,
        param_distributions=params,
        n_iter=N_ITER_SEARCH,
        scoring="roc_auc",
        cv=inner_cv,
        n_jobs=-1,
        verbose=0,
        random_state=RANDOM_STATE,
        error_score=np.nan,
        refit=True,
    )
    search.fit(X, y)
    return search


def proba_positive(model: Any, X: Any) -> Any:
    import numpy as np
    if hasattr(model, "predict_proba"):
        p = model.predict_proba(X)
        if p.ndim == 2 and p.shape[1] > 1:
            return p[:, 1]
        return p.reshape(-1)
    if hasattr(model, "decision_function"):
        z = model.decision_function(X)
        return 1.0 / (1.0 + np.exp(-z))
    return model.predict(X).astype(float)


def metrics_from_predictions(y_true: Any, y_pred: Any, y_prob: Any) -> Dict[str, Any]:
    import numpy as np
    from sklearn.metrics import (
        accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
        log_loss, confusion_matrix, brier_score_loss,
    )

    y_true = np.asarray(y_true).astype(int)
    y_pred = np.asarray(y_pred).astype(int)
    y_prob = np.clip(np.asarray(y_prob).astype(float), 1e-7, 1 - 1e-7)
    labels = [0, 1]
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=labels).ravel()
    specificity = tn / (tn + fp) if (tn + fp) > 0 else np.nan
    sensitivity = tp / (tp + fn) if (tp + fn) > 0 else np.nan
    ppv = tp / (tp + fp) if (tp + fp) > 0 else np.nan
    npv = tn / (tn + fn) if (tn + fn) > 0 else np.nan
    try:
        auc = roc_auc_score(y_true, y_prob)
    except Exception:
        auc = np.nan
    try:
        ll = log_loss(y_true, y_prob, labels=labels)
    except Exception:
        ll = np.nan
    try:
        brier = brier_score_loss(y_true, y_prob)
    except Exception:
        brier = np.nan
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Sensitivity_Recall": sensitivity,
        "Specificity": specificity,
        "PPV_Precision": ppv,
        "NPV": npv,
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "ROC_AUC": auc,
        "Log_Loss": ll,
        "Brier": brier,
        "TN": int(tn),
        "FP": int(fp),
        "FN": int(fn),
        "TP": int(tp),
    }


def bootstrap_metric_ci(y_true: Any, y_pred: Any, y_prob: Any, n_boot: int = BOOTSTRAP_N) -> Any:
    import numpy as np
    import pandas as pd

    rng = np.random.default_rng(BOOTSTRAP_SEED)
    y_true = np.asarray(y_true).astype(int)
    y_pred = np.asarray(y_pred).astype(int)
    y_prob = np.asarray(y_prob).astype(float)
    n = len(y_true)
    rows = []
    for _ in range(n_boot):
        idx = rng.integers(0, n, n)
        if len(np.unique(y_true[idx])) < 2:
            continue
        rows.append(metrics_from_predictions(y_true[idx], y_pred[idx], y_prob[idx]))
    boot = pd.DataFrame(rows)
    ci_rows = []
    alpha = 1.0 - float(BOOTSTRAP_METRIC_CI_LEVEL)
    low_q = 100.0 * alpha / 2.0
    high_q = 100.0 * (1.0 - alpha / 2.0)
    for metric in ["Accuracy", "Sensitivity_Recall", "Specificity", "PPV_Precision", "NPV", "F1", "ROC_AUC", "Log_Loss", "Brier"]:
        vals = boot[metric].dropna().values if metric in boot else []
        if len(vals) == 0:
            ci_rows.append({"Metric": metric, "CI_level": BOOTSTRAP_METRIC_CI_LEVEL, "CI_low": np.nan, "CI_high": np.nan})
        else:
            ci_rows.append({"Metric": metric, "CI_level": BOOTSTRAP_METRIC_CI_LEVEL, "CI_low": np.percentile(vals, low_q), "CI_high": np.percentile(vals, high_q)})
    return pd.DataFrame(ci_rows)


def calibration_intercept_slope(y_true: Any, y_prob: Any) -> Dict[str, float]:
    import numpy as np
    from sklearn.linear_model import LogisticRegression

    y = np.asarray(y_true).astype(int)
    p = np.clip(np.asarray(y_prob).astype(float), 1e-6, 1 - 1e-6)
    logit_p = np.log(p / (1 - p)).reshape(-1, 1)
    try:
        cal = LogisticRegression(penalty=None, solver="lbfgs", max_iter=2000)
        cal.fit(logit_p, y)
        return {"calibration_intercept": float(cal.intercept_[0]), "calibration_slope": float(cal.coef_[0][0])}
    except Exception:
        try:
            cal = LogisticRegression(solver="lbfgs", C=1e6, max_iter=2000)
            cal.fit(logit_p, y)
            return {"calibration_intercept": float(cal.intercept_[0]), "calibration_slope": float(cal.coef_[0][0])}
        except Exception:
            return {"calibration_intercept": np.nan, "calibration_slope": np.nan}


def plot_roc_comparison(y_cv: Any, p_cv: Any, y_test: Any, p_test: Any, title: str, path: Path) -> None:
    import numpy as np
    import matplotlib.pyplot as plt
    from sklearn.metrics import roc_curve, roc_auc_score

    fig, ax = plt.subplots(figsize=(8, 6))
    if len(np.unique(y_cv)) == 2:
        fpr, tpr, _ = roc_curve(y_cv, p_cv)
        auc = roc_auc_score(y_cv, p_cv)
        ax.plot(fpr, tpr, lw=2, label=f"Nested CV outer predictions (AUC = {auc:.4f})")
    if len(np.unique(y_test)) == 2:
        fpr, tpr, _ = roc_curve(y_test, p_test)
        auc = roc_auc_score(y_test, p_test)
        ax.plot(fpr, tpr, lw=2, label=f"Held-out test n={len(y_test)} (AUC = {auc:.4f})")
    ax.plot([0, 1], [0, 1], "k--")
    ax.set_xlabel("False Positive Rate")
    ax.set_ylabel("True Positive Rate")
    ax.set_title(title, fontsize=14, fontweight="bold")
    ax.legend(loc="lower right")
    ax.grid(True, alpha=0.4)
    fig.tight_layout()
    save_or_show(fig, path)


def plot_confusion_matrices(y_cv: Any, pred_cv: Any, y_test: Any, pred_test: Any, title: str, path: Path) -> None:
    import matplotlib.pyplot as plt
    from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

    labels = ["Benign ovarian tumor", "Epithelial ovarian cancer"]
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    ConfusionMatrixDisplay(confusion_matrix(y_cv, pred_cv, labels=[0, 1]), display_labels=labels).plot(
        ax=axes[0], cmap="Blues", colorbar=False, values_format="d"
    )
    axes[0].set_title("Nested CV outer predictions")
    ConfusionMatrixDisplay(confusion_matrix(y_test, pred_test, labels=[0, 1]), display_labels=labels).plot(
        ax=axes[1], cmap="Blues", colorbar=False, values_format="d"
    )
    axes[1].set_title(f"Held-out test n={len(y_test)}")
    fig.suptitle(title, fontsize=14, fontweight="bold")
    fig.tight_layout()
    save_or_show(fig, path)



def calibration_bins_quantile(y_true: Any, y_prob: Any, n_bins: int = CALIBRATION_N_BINS) -> Any:
    import pandas as pd
    y = np.asarray(y_true).astype(int)
    p = np.asarray(y_prob).astype(float)
    df = pd.DataFrame({"y": y, "p": p})
    try:
        df["bin"] = pd.qcut(df["p"], q=min(n_bins, len(df)), duplicates="drop")
    except Exception:
        df["bin"] = pd.cut(df["p"], bins=min(n_bins, max(3, len(df) // 10)), include_lowest=True)
    rows = []
    for _, g in df.groupby("bin", observed=False):
        n = int(len(g))
        k = int(g["y"].sum())
        lo, hi = wilson_ci(k, n)
        obs = float(k / n) if n else np.nan
        mean_p = float(g["p"].mean()) if n else np.nan
        rows.append({
            "mean_predicted_probability": mean_p,
            "observed_event_rate": obs,
            "n": n,
            "events": k,
            "wilson_low": lo,
            "wilson_high": hi,
            "err_low": obs - lo if n else np.nan,
            "err_high": hi - obs if n else np.nan,
        })
    return pd.DataFrame(rows).sort_values("mean_predicted_probability")


def bootstrap_calibration_ci(y_true: Any, y_prob: Any, n_boot: int = BOOTSTRAP_N) -> Dict[str, Tuple[float, float]]:
    rng = np.random.default_rng(BOOTSTRAP_SEED)
    y = np.asarray(y_true).astype(int)
    p = np.clip(np.asarray(y_prob).astype(float), 1e-6, 1 - 1e-6)
    vals = {"Brier": [], "calibration_intercept": [], "calibration_slope": []}
    for _ in range(n_boot):
        idx = rng.integers(0, len(y), len(y))
        if len(np.unique(y[idx])) < 2:
            continue
        vals["Brier"].append(float(((p[idx] - y[idx]) ** 2).mean()))
        cs = calibration_intercept_slope(y[idx], p[idx])
        vals["calibration_intercept"].append(cs.get("calibration_intercept", np.nan))
        vals["calibration_slope"].append(cs.get("calibration_slope", np.nan))
    out: Dict[str, Tuple[float, float]] = {}
    for k, arr in vals.items():
        a = np.asarray(arr, dtype=float)
        a = a[np.isfinite(a)]
        if len(a) == 0:
            out[k] = (np.nan, np.nan)
        else:
            out[k] = (float(np.percentile(a, 2.5)), float(np.percentile(a, 97.5)))
    return out


def plot_calibration(y_true: Any, y_prob: Any, title: str, path: Path) -> Dict[str, float]:
    import matplotlib.pyplot as plt
    from matplotlib.ticker import MultipleLocator
    from sklearn.metrics import brier_score_loss

    y = np.asarray(y_true).astype(int)
    p = np.clip(np.asarray(y_prob).astype(float), 1e-7, 1 - 1e-7)
    bins = calibration_bins_quantile(y, p, CALIBRATION_N_BINS)
    brier = float(brier_score_loss(y, p))
    cal = calibration_intercept_slope(y, p)
    ci = bootstrap_calibration_ci(y, p, BOOTSTRAP_N)

    fig, ax = plt.subplots(figsize=(9, 7))
    fig.patch.set_facecolor("#F8FAFC")
    ax.set_facecolor("white")

    x = bins["mean_predicted_probability"].astype(float).values
    obs = bins["observed_event_rate"].astype(float).values
    lo = bins["wilson_low"].astype(float).values
    hi = bins["wilson_high"].astype(float).values

    # Matplotlib rejects negative yerr. Wilson bounds should contain the observed rate,
    # but floating-point noise or duplicated quantile edges can create tiny negatives.
    lo = np.clip(np.minimum(lo, obs), 0.0, 1.0)
    hi = np.clip(np.maximum(hi, obs), 0.0, 1.0)
    err_low = np.maximum(0.0, obs - lo)
    err_high = np.maximum(0.0, hi - obs)

    ax.plot([0, 1], [0, 1], "--", color="#94A3B8", linewidth=1.8, alpha=0.85,
            label="Perfect calibration", zorder=1)
    ax.fill_between(x, lo, hi, color="#BAE6FD", alpha=0.45,
                    label="Wilson 95% CI band", zorder=2)
    ax.plot(x, obs, color="#0284C7", linewidth=2.7, marker="o", markersize=7,
            markerfacecolor="#38BDF8", markeredgewidth=1.6, markeredgecolor="#0284C7",
            label="Observed rate per bin", zorder=4)
    ax.errorbar(x, obs, yerr=[err_low, err_high], fmt="none", ecolor="#0284C7",
                elinewidth=1.4, capsize=5, capthick=1.4, alpha=0.85, zorder=3)

    used_label_positions = []
    for _, r in bins.iterrows():
        xp = float(r["mean_predicted_probability"])
        yp = float(r["observed_event_rate"])
        if np.isfinite(xp) and np.isfinite(yp):
            ytxt = min(1.10, max(0.04, yp + 0.08))
            for old_x, old_y in used_label_positions:
                if abs(old_x - xp) < 0.08 and abs(old_y - ytxt) < 0.08:
                    ytxt = max(0.04, yp - 0.10)
                    break
            used_label_positions.append((xp, ytxt))
            ax.annotate(
                f"n={int(r['n'])}",
                xy=(xp, yp), xytext=(xp, ytxt),
                ha="center", va="center", fontsize=9, color="#334155", fontweight="bold",
                arrowprops=dict(arrowstyle="-", color="#CBD5E1", lw=0.9),
                bbox=dict(boxstyle="round,pad=0.25", fc="white", ec="#CBD5E1", alpha=0.95),
                annotation_clip=False,
            )

    ax.set_xlim(-0.02, 1.02)
    ax.set_ylim(-0.02, 1.16)
    ax.set_xlabel("Mean Predicted Probability", fontsize=12)
    ax.set_ylabel("Observed Event Rate", fontsize=12)
    ax.set_title(
        f"{title}\nn = {len(y)}   |   {CALIBRATION_N_BINS}-bin quantile   |   Wilson 95% CI",
        fontsize=15, fontweight="bold", color="#0F172A", pad=12,
    )
    ax.xaxis.set_minor_locator(MultipleLocator(0.05))
    ax.yaxis.set_minor_locator(MultipleLocator(0.05))
    ax.grid(True, which="major", color="#E2E8F0", linewidth=0.8)
    ax.grid(True, which="minor", color="#E2E8F0", linewidth=0.35, alpha=0.55)

    metric_text = (
        "Calibration Metrics\n"
        "Estimate        95% CI\n"
        f"Brier Score   {brier:0.4f}   [{ci['Brier'][0]:0.4f}, {ci['Brier'][1]:0.4f}]\n"
        f"Intercept     {cal['calibration_intercept']:0.4f}   [{ci['calibration_intercept'][0]:0.4f}, {ci['calibration_intercept'][1]:0.4f}]\n"
        f"Slope         {cal['calibration_slope']:0.4f}   [{ci['calibration_slope'][0]:0.4f}, {ci['calibration_slope'][1]:0.4f}]"
    )
    ax.text(
        0.055, 0.92, metric_text, transform=ax.transAxes, va="top", ha="left", fontsize=9, color="#334155",
        bbox=dict(boxstyle="round,pad=0.60", fc="white", ec="#CBD5E1", alpha=0.93),
        zorder=10,
    )
    ax.legend(loc="lower right", framealpha=0.95)
    for sp in ["top", "right"]:
        ax.spines[sp].set_visible(False)
    fig.tight_layout()
    save_or_show(fig, path)

    out = {"Brier": brier, **cal}
    out.update({f"{k}_ci_low": v[0] for k, v in ci.items()})
    out.update({f"{k}_ci_high": v[1] for k, v in ci.items()})
    bins.to_csv(path.with_suffix(".bins.csv"), index=False)
    pd.DataFrame([out]).to_csv(path.with_suffix(".metrics.csv"), index=False)
    return out

def plot_predicted_probability_distribution(y_true: Any, y_prob: Any, title: str, path: Path) -> None:
    import matplotlib.pyplot as plt
    y = np.asarray(y_true).astype(int)
    p = np.clip(np.asarray(y_prob).astype(float), 0, 1)
    p0 = p[y == 0]
    p1 = p[y == 1]
    fig, ax = plt.subplots(figsize=(9, 6))
    ax.hist(p0, bins=15, alpha=0.70, label="True class 0")
    ax.hist(p1, bins=15, alpha=0.70, label="True class 1")
    if len(p0):
        m0 = float(np.median(p0))
        ax.axvline(m0, linestyle="--", linewidth=2, label=f"Median class 0 = {m0:.2f}")
    if len(p1):
        m1 = float(np.median(p1))
        ax.axvline(m1, linestyle="--", linewidth=2, label=f"Median class 1 = {m1:.2f}")
    ax.set_xlim(0, 1)
    ax.set_xlabel("Predicted Probability")
    ax.set_ylabel("Count")
    ax.set_title(f"{title}\nHeld-out class separation between true negatives and positives", fontsize=15, fontweight="bold")
    ax.grid(True, alpha=0.25)
    ax.legend(loc="best")
    fig.tight_layout()
    save_or_show(fig, path)


def plot_selection_stability(stability_df: Any, selector: str, out_dirs: Dict[str, Path]) -> None:
    import matplotlib.pyplot as plt
    import seaborn as sns

    if stability_df.empty:
        return
    df = stability_df.sort_values("selection_frequency", ascending=False).head(30)
    fig, ax = plt.subplots(figsize=(12, max(6, 0.35 * len(df))))
    sns.barplot(data=df, x="selection_frequency", y="feature", ax=ax)
    ax.set_title(f"{selector}: Feature Selection Stability Across Outer Folds", fontsize=14, fontweight="bold")
    ax.set_xlabel(f"Frequency out of {OUTER_SPLITS} outer folds")
    ax.set_ylabel("Feature")
    fig.tight_layout()
    save_or_show(fig, out_dirs["figures"] / f"{selector}_selection_stability.png")


def nested_cv_evaluate_selector(
    selector_name: str,
    X_train_raw: Any,
    y_train: Any,
    out_dirs: Dict[str, Path],
    representation: str = "binned",
) -> Tuple[Any, Dict[str, Any], Any]:
    import numpy as np
    import pandas as pd
    from sklearn.model_selection import StratifiedKFold

    model_defs = model_spaces()
    outer_cv = StratifiedKFold(n_splits=OUTER_SPLITS, shuffle=True, random_state=RANDOM_STATE)
    fold_metric_rows = []
    oof_store = {model_name: {"y_true": [], "y_pred": [], "y_prob": []} for model_name in model_defs}
    selected_by_fold: List[List[str]] = []
    selector_info_by_fold = []

    X_train_raw, y_train = reset_xy(X_train_raw, y_train)

    for fold, (tr_idx, val_idx) in enumerate(outer_cv.split(X_train_raw, y_train), start=1):
        print(f"\n===== {selector_name} | Outer fold {fold}/{OUTER_SPLITS} =====")
        X_tr_raw = X_train_raw.iloc[tr_idx].reset_index(drop=True)
        X_val_raw = X_train_raw.iloc[val_idx].reset_index(drop=True)
        y_tr = y_train.iloc[tr_idx].reset_index(drop=True)
        y_val = y_train.iloc[val_idx].reset_index(drop=True)

        fold_label = f"{selector_name}_outer_fold_{fold}"
        prep, selected_features, selector_info, X_tr_prep, X_filtered, _, _, _ = run_fold_preprocess_and_select(
            selector_name, X_tr_raw, y_tr, out_dirs, fold_label, representation=representation
        )
        selected_by_fold.append(selected_features)
        selector_info_by_fold.append(selector_info)
        print_selector_details(f"{selector_name} outer fold {fold}", selector_info)
        print(f"{selector_name} outer fold {fold} selected features ({len(selected_features)}): {selected_features}")
        X_val_prep = prep.transform(X_val_raw).reset_index(drop=True)

        X_tr_sel = X_tr_prep[selected_features]
        X_val_sel = X_val_prep[selected_features]

        for model_name, spec in model_defs.items():
            print(f"Tuning/evaluating {selector_name} | fold {fold} | model {model_name}")
            try:
                search = tune_model(spec["estimator"], spec["params"], X_tr_sel, y_tr)
                best_model = search.best_estimator_
                p_val = proba_positive(best_model, X_val_sel)
                pred_val = (p_val >= 0.5).astype(int)
                met = metrics_from_predictions(y_val, pred_val, p_val)
                row = {
                    "Selector": selector_name,
                    "Representation": representation,
                    "Fold": fold,
                    "Model": model_name,
                    "Num_Selected_Features": len(selected_features),
                    "Selected_Features": ";".join(selected_features),
                    "Best_Params": json.dumps(search.best_params_, default=str),
                    **met,
                }
                fold_metric_rows.append(row)
                oof_store[model_name]["y_true"].extend(y_val.tolist())
                oof_store[model_name]["y_pred"].extend(pred_val.tolist())
                oof_store[model_name]["y_prob"].extend(p_val.tolist())
                metric_line(f"OUTER FOLD RESULT | {selector_name} | fold={fold} | model={model_name}", row)
            except Exception as e:
                print(f"Model failed: {selector_name} fold {fold} model {model_name}: {e}")
                fold_metric_rows.append({
                    "Selector": selector_name,
                    "Representation": representation,
                    "Fold": fold,
                    "Model": model_name,
                    "Num_Selected_Features": len(selected_features),
                    "Selected_Features": ";".join(selected_features),
                    "Error": str(e),
                })

    fold_df = pd.DataFrame(fold_metric_rows)
    fold_df.to_csv(out_dirs["tables"] / f"{selector_name}_nested_cv_fold_metrics.csv", index=False)

    # Mean/std summary.
    metric_cols = ["Accuracy", "Sensitivity_Recall", "Specificity", "PPV_Precision", "NPV", "F1", "ROC_AUC", "Log_Loss", "Brier"]
    summary_rows = []
    for model_name in model_defs:
        sub = fold_df[fold_df["Model"] == model_name]
        row = {"Selector": selector_name, "Representation": representation, "Model": model_name}
        for m in metric_cols:
            if m in sub:
                row[f"{m}_mean"] = pd.to_numeric(sub[m], errors="coerce").mean()
                row[f"{m}_std"] = pd.to_numeric(sub[m], errors="coerce").std()
        # Pooled outer predictions.
        if len(oof_store[model_name]["y_true"]) > 0:
            pooled = metrics_from_predictions(
                oof_store[model_name]["y_true"], oof_store[model_name]["y_pred"], oof_store[model_name]["y_prob"]
            )
            for k, v in pooled.items():
                row[f"Pooled_{k}"] = v
        summary_rows.append(row)
    summary_df = pd.DataFrame(summary_rows)
    summary_df.to_csv(out_dirs["tables"] / f"{selector_name}_nested_cv_summary.csv", index=False)

    metric_cols_print = [c for c in ["Selector", "Fold", "Model", "Num_Selected_Features", "Accuracy", "Sensitivity_Recall", "Specificity", "PPV_Precision", "NPV", "F1", "ROC_AUC", "Log_Loss", "Brier", "TN", "FP", "FN", "TP", "Selected_Features"] if c in fold_df.columns]
    print_df(f"{selector_name}: ALL OUTER-FOLD RESULTS FOR ALL MODELS", fold_df[metric_cols_print], max_rows=None)
    print_df(f"{selector_name}: NESTED-CV SUMMARY FOR ALL MODELS", summary_df, max_rows=None)

    all_selected = [f for fold_feats in selected_by_fold for f in fold_feats]
    stability = pd.Series(all_selected).value_counts().rename_axis("feature").reset_index(name="selection_frequency")
    stability["selection_rate"] = stability["selection_frequency"] / OUTER_SPLITS
    stability.to_csv(out_dirs["features"] / f"{selector_name}_selection_stability.csv", index=False)
    print_df(f"{selector_name}: FEATURE-SELECTION STABILITY ACROSS OUTER FOLDS", stability, max_rows=None)
    plot_selection_stability(stability, selector_name, out_dirs)

    return fold_df, oof_store, stability



def _save_current_figure(path: Path) -> None:
    fig = plt.gcf()
    fig.savefig(path, dpi=300, bbox_inches="tight", facecolor="white")
    if SHOW_FIGS or DISPLAY_PER_FIGURE_AS_CREATED:
        plt.show()
    else:
        plt.close(fig)


def _beautiful_horizontal_bar(labels: List[str], values: List[float], title: str, xlabel: str, path: Path, signed: bool = False) -> None:
    vals = np.asarray(values, dtype=float)
    order = np.argsort(np.abs(vals) if signed else vals)
    labels_ord = [labels[i] for i in order]
    vals_ord = vals[order]
    fig, ax = plt.subplots(figsize=(10, max(6, 0.42 * len(labels_ord))))
    colors = ["#E11D48" if v >= 0 else "#0284C7" for v in vals_ord] if signed else ["#0284C7"] * len(vals_ord)
    ax.barh(labels_ord, vals_ord, color=colors, edgecolor="white", linewidth=0.8)
    ax.axvline(0, color="#94A3B8", linewidth=1.1)
    ax.set_title(title, fontsize=15, fontweight="bold", color="#0F172A", pad=12)
    ax.set_xlabel(xlabel, fontsize=12, color="#334155")
    ax.grid(True, axis="x", color="#E2E8F0", alpha=0.9)
    ax.set_facecolor("white")
    fig.patch.set_facecolor("#F8FAFC")
    for sp in ["top", "right"]:
        ax.spines[sp].set_visible(False)
    fig.tight_layout()
    save_or_show(fig, path)


def explain_best_model_with_xai(
    selector_name: str,
    model_name: str,
    best_model: Any,
    X_train_sel: Any,
    X_test_sel: Any,
    y_test: Any,
    selected_features: List[str],
    out_dirs: Dict[str, Path],
) -> None:
    if selector_name != XAI_FOCUS_SELECTOR:
        print(f"XAI skipped for {selector_name}. Main XAI focus is {XAI_FOCUS_SELECTOR}.")
        return

    print_block(f"MAIN XAI FOR {selector_name} + {model_name}")
    try:
        import shap
        import lime.lime_tabular
    except Exception as e:
        print(f"XAI packages unavailable: {e}")
        return

    xai_dir = out_dirs["figures"] / "xai"
    xai_dir.mkdir(parents=True, exist_ok=True)
    table_dir = out_dirs["tables"] / "xai"
    table_dir.mkdir(parents=True, exist_ok=True)

    X_train_sel = X_train_sel.copy().reset_index(drop=True)
    X_test_sel = X_test_sel.copy().reset_index(drop=True)
    y_arr = np.asarray(y_test).astype(int)

    bg = X_train_sel
    if len(bg) > XAI_BACKGROUND_SIZE:
        bg = bg.sample(n=XAI_BACKGROUND_SIZE, random_state=RANDOM_STATE).reset_index(drop=True)
    eval_X = X_test_sel
    if len(eval_X) > XAI_EVAL_SIZE:
        eval_X = eval_X.sample(n=XAI_EVAL_SIZE, random_state=RANDOM_STATE).reset_index(drop=True)

    test_prob = np.asarray(proba_positive(best_model, X_test_sel), dtype=float)
    positive_candidates = np.where(y_arr == 1)[0]
    negative_candidates = np.where(y_arr == 0)[0]
    pos_idx = int(positive_candidates[np.argmax(test_prob[positive_candidates])]) if len(positive_candidates) else int(np.argmax(test_prob))
    neg_idx = int(negative_candidates[np.argmin(test_prob[negative_candidates])]) if len(negative_candidates) else int(np.argmin(test_prob))

    def model_fn(x):
        X_df = pd.DataFrame(x, columns=selected_features)
        return np.asarray(proba_positive(best_model, X_df), dtype=float)

    try:
        masker = shap.maskers.Independent(bg)
        explainer = shap.Explainer(model_fn, masker, feature_names=selected_features)
        shap_values = explainer(eval_X)
        values = np.asarray(shap_values.values, dtype=float)
        if values.ndim == 3:
            values = values[:, :, -1]
        mean_abs = np.abs(values).mean(axis=0)
        shap_imp = pd.DataFrame({"feature": selected_features, "mean_abs_shap": mean_abs}).sort_values("mean_abs_shap", ascending=False)
        shap_imp.to_csv(table_dir / f"{selector_name}_{model_name}_main_xai_shap_importance.csv", index=False)
        print_df(f"{selector_name} + {model_name}: XAI SHAP IMPORTANCE", shap_imp, max_rows=None)

        top = shap_imp.head(min(XAI_MAX_DISPLAY_FEATURES, len(shap_imp)))
        _beautiful_horizontal_bar(
            top["feature"].tolist()[::-1],
            top["mean_abs_shap"].tolist()[::-1],
            f"{selector_name} + {model_name} · SHAP Bar Plot",
            "mean(|SHAP value|): average impact on predicted risk",
            xai_dir / f"{selector_name}_{model_name}_XAI_01_shap_bar.png",
            signed=False,
        )

        plt.figure(figsize=(10, max(7, 0.38 * min(XAI_MAX_DISPLAY_FEATURES, len(selected_features)))))
        shap.plots.beeswarm(shap_values, max_display=min(XAI_MAX_DISPLAY_FEATURES, len(selected_features)), show=False)
        plt.title(f"{selector_name} + {model_name} · SHAP Beeswarm", fontsize=15, fontweight="bold", pad=12)
        plt.tight_layout()
        _save_current_figure(xai_dir / f"{selector_name}_{model_name}_XAI_02_shap_beeswarm.png")

        eval_pos_idx = pos_idx if pos_idx < len(eval_X) else 0
        plt.figure(figsize=(13, 4.8))
        try:
            shap.plots.force(shap_values[eval_pos_idx], matplotlib=True, show=False)
            plt.title(f"{selector_name} + {model_name} · SHAP Force Plot: high-risk held-out case", fontsize=13, fontweight="bold")
            plt.tight_layout()
            _save_current_figure(xai_dir / f"{selector_name}_{model_name}_XAI_03_shap_force_positive_case.png")
        except Exception as e:
            plt.close()
            print(f"Matplotlib SHAP force plot skipped: {e}")
            try:
                force_html = shap.plots.force(shap_values[eval_pos_idx], matplotlib=False)
                shap.save_html(str(xai_dir / f"{selector_name}_{model_name}_XAI_03_shap_force_positive_case.html"), force_html)
            except Exception as ee:
                print(f"HTML SHAP force plot also skipped: {ee}")

        plt.figure(figsize=(10, 7))
        shap.plots.waterfall(shap_values[eval_pos_idx], max_display=min(14, len(selected_features)), show=False)
        plt.title(f"{selector_name} + {model_name} · SHAP Waterfall: high-risk held-out case", fontsize=14, fontweight="bold")
        plt.tight_layout()
        _save_current_figure(xai_dir / f"{selector_name}_{model_name}_XAI_04_shap_waterfall_positive_case.png")

        if len(top) > 0:
            for rank, feat in enumerate(top["feature"].head(min(3, len(top))).tolist(), start=1):
                try:
                    plt.figure(figsize=(8, 6))
                    shap.plots.scatter(shap_values[:, feat], color=shap_values, show=False)
                    plt.title(f"{selector_name} + {model_name} · SHAP Dependence: {feat}", fontsize=14, fontweight="bold")
                    plt.tight_layout()
                    _save_current_figure(xai_dir / f"{selector_name}_{model_name}_XAI_extra_dependence_{rank}_{feat.replace('#','num').replace('%','pct')}.png")
                except Exception as e:
                    print(f"SHAP dependence skipped for {feat}: {e}")
    except Exception as e:
        print(f"SHAP explanation failed for {selector_name} + {model_name}: {e}")

    try:
        class_names = ["Benign ovarian tumor", "Epithelial ovarian cancer"]
        lime_explainer = lime.lime_tabular.LimeTabularExplainer(
            training_data=np.asarray(X_train_sel, dtype=float),
            feature_names=selected_features,
            class_names=class_names,
            mode="classification",
            discretize_continuous=False,
            random_state=RANDOM_STATE,
        )

        def lime_predict(x):
            X_df = pd.DataFrame(x, columns=selected_features)
            p = np.asarray(proba_positive(best_model, X_df), dtype=float)
            return np.vstack([1 - p, p]).T

        for label_name, idx, filename_suffix in [
            ("high-risk ovarian-cancer held-out case", pos_idx, "positive_case"),
            ("low-risk non-cancer held-out case", neg_idx, "negative_case"),
        ]:
            exp = lime_explainer.explain_instance(
                np.asarray(X_test_sel.iloc[idx], dtype=float),
                lime_predict,
                num_features=min(10, len(selected_features)),
                labels=[1],
            )
            html_path = xai_dir / f"{selector_name}_{model_name}_XAI_lime_{filename_suffix}.html"
            exp.save_to_file(str(html_path))
            exp_list = exp.as_list(label=1)
            lime_df = pd.DataFrame(exp_list, columns=["feature_rule", "contribution_to_class_1"])
            lime_df.to_csv(table_dir / f"{selector_name}_{model_name}_XAI_lime_{filename_suffix}.csv", index=False)
            _beautiful_horizontal_bar(
                lime_df["feature_rule"].tolist()[::-1],
                lime_df["contribution_to_class_1"].astype(float).tolist()[::-1],
                f"{selector_name} + {model_name} · LIME: {label_name}",
                "Contribution toward ovarian-cancer probability",
                xai_dir / f"{selector_name}_{model_name}_XAI_05_lime_{filename_suffix}.png" if filename_suffix == "positive_case" else xai_dir / f"{selector_name}_{model_name}_XAI_06_lime_{filename_suffix}.png",
                signed=True,
            )
    except Exception as e:
        print(f"LIME explanation skipped for {selector_name} + {model_name}: {e}")

def run_lr_baseline_comparison(
    selector_name: str,
    X_train_sel: Any,
    y_train: Any,
    X_test_sel: Any,
    y_test: Any,
    out_dirs: Dict[str, Path],
) -> Any:
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.linear_model import LogisticRegression

    if not RUN_LR_BASELINES_FOR_EACH_SELECTOR:
        return pd.DataFrame()

    baselines = {
        "Standard LR": LogisticRegression(penalty=None, solver="lbfgs", max_iter=5000, random_state=RANDOM_STATE),
        "Ridge LR": LogisticRegression(penalty="l2", C=1.0, solver="lbfgs", max_iter=5000, random_state=RANDOM_STATE),
        "LASSO LR": LogisticRegression(penalty="l1", C=1.0, solver="saga", max_iter=7000, random_state=RANDOM_STATE),
        "Elastic-Net LR": LogisticRegression(penalty="elasticnet", C=1.0, l1_ratio=0.5, solver="saga", max_iter=7000, random_state=RANDOM_STATE),
    }
    rows = []
    for name, lr in baselines.items():
        model = Pipeline([("scaler", StandardScaler()), ("lr", lr)])
        try:
            model.fit(X_train_sel, y_train)
            p = proba_positive(model, X_test_sel)
            pred = (p >= 0.5).astype(int)
            row = {"Selector": selector_name, "Baseline": name, "Feature_Set": f"{selector_name}_selected_features", "Num_Selected_Features": X_train_sel.shape[1], **metrics_from_predictions(y_test, pred, p), **calibration_intercept_slope(y_test, p)}
            rows.append(row)
            metric_line(f"LR BASELINE | {selector_name} | {name}", row)
        except Exception as e:
            rows.append({"Selector": selector_name, "Baseline": name, "Error": str(e)})
            print(f"LR baseline failed | {selector_name} | {name}: {e}")
    df = pd.DataFrame(rows)
    df.to_csv(out_dirs["tables"] / f"{selector_name}_LR_baseline_comparison_on_selected_features.csv", index=False)
    print_df(f"{selector_name}: EMPIRICAL LR BASELINE COMPARISON ON {selector_name} SELECTED FEATURES", df, max_rows=None)
    return df




def run_stat_aq_xgb_variant_baselines(
    selector_name: str,
    X_train_sel: Any,
    y_train: Any,
    X_test_sel: Any,
    y_test: Any,
    out_dirs: Dict[str, Path],
    tuned_best_model: Optional[Any] = None,
) -> Any:
    if selector_name != "Stat-AQ" or not RUN_STAT_AQ_XGB_VARIANT_BASELINES:
        return pd.DataFrame()
    runtime = import_runtime_packages()
    XGBClassifier = runtime.get("XGBClassifier")
    if XGBClassifier is None:
        print("XGB variant baselines skipped because xgboost is unavailable.")
        return pd.DataFrame()

    print("Stat-AQ XGB baseline comparison uses weaker/default XGB profiles only; the tuned selected base model is intentionally excluded from this baseline table.")

    variants: Dict[str, Any] = {}
    variants["Default XGB no tuning"] = XGBClassifier(
        eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1
    )
    variants["Legacy-style XGB stumps 25 trees"] = XGBClassifier(
        n_estimators=25, max_depth=1, learning_rate=0.30,
        subsample=1.00, colsample_bytree=1.00, min_child_weight=1,
        reg_alpha=0.0, reg_lambda=1.0, gamma=0.0,
        eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1,
    )
    variants["Very weak XGB 10 trees"] = XGBClassifier(
        n_estimators=10, max_depth=2, learning_rate=0.30,
        subsample=1.00, colsample_bytree=1.00, min_child_weight=1,
        reg_alpha=0.0, reg_lambda=1.0, gamma=0.0,
        eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1,
    )
    variants["Conservative XGB depth2"] = XGBClassifier(
        n_estimators=120, max_depth=2, learning_rate=0.05,
        subsample=0.85, colsample_bytree=0.85, min_child_weight=3,
        eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1,
    )
    variants["Regularized XGB"] = XGBClassifier(
        n_estimators=200, max_depth=3, learning_rate=0.05,
        subsample=0.80, colsample_bytree=0.80, min_child_weight=2,
        reg_alpha=0.5, reg_lambda=2.0, gamma=0.1,
        eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1,
    )
    variants["Parsimonious XGB 50 trees"] = XGBClassifier(
        n_estimators=50, max_depth=2, learning_rate=0.10,
        subsample=0.90, colsample_bytree=0.90,
        eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1,
    )
    variants["Extra conservative XGB stumps 75 trees"] = XGBClassifier(
        n_estimators=75, max_depth=1, learning_rate=0.05,
        subsample=0.80, colsample_bytree=0.80, min_child_weight=5,
        reg_alpha=0.25, reg_lambda=3.0, gamma=0.20,
        eval_metric="logloss", random_state=RANDOM_STATE, n_jobs=-1,
    )

    rows = []
    ci_tables = []
    for name, model in variants.items():
        try:
            model.fit(X_train_sel, y_train)
            p = proba_positive(model, X_test_sel)
            pred = (p >= 0.5).astype(int)
            row = {
                "Selector": selector_name,
                "Baseline_Group": "Stat-AQ XGB variant baseline",
                "Variant": name,
                "Num_Selected_Features": X_train_sel.shape[1],
                **metrics_from_predictions(y_test, pred, p),
                **calibration_intercept_slope(y_test, p),
            }
            rows.append(row)
            metric_line(f"XGB VARIANT BASELINE | {selector_name} | {name}", row)
            ci = bootstrap_metric_ci(y_test, pred, p)
            ci.insert(0, "Variant", name)
            ci.insert(0, "Selector", selector_name)
            ci_tables.append(ci)
        except Exception as e:
            rows.append({"Selector": selector_name, "Variant": name, "Error": str(e)})
            print(f"XGB variant baseline failed | {selector_name} | {name}: {e}")

    df = pd.DataFrame(rows)
    df = df.sort_values(["ROC_AUC", "Log_Loss", "F1"], ascending=[False, True, False], na_position="last") if "ROC_AUC" in df.columns else df
    df.to_csv(out_dirs["tables"] / f"{selector_name}_XGB_variant_baseline_comparison.csv", index=False)
    print_df(f"{selector_name}: XGB VARIANT BASELINE COMPARISON", df, max_rows=None)
    if ci_tables:
        ci_df = pd.concat(ci_tables, ignore_index=True)
        ci_df.to_csv(out_dirs["tables"] / f"{selector_name}_XGB_variant_baseline_bootstrap_{int(BOOTSTRAP_METRIC_CI_LEVEL*100)}ci.csv", index=False)
        print_df(f"{selector_name}: XGB VARIANT BASELINE BOOTSTRAP {int(BOOTSTRAP_METRIC_CI_LEVEL*100)}% CI", ci_df, max_rows=None)
    return df


def run_clinical_biomarker_baselines(
    X_train_raw: Any,
    y_train: Any,
    X_test_raw: Any,
    y_test: Any,
    out_dirs: Dict[str, Path],
) -> Any:
    if not RUN_CLINICAL_BIOMARKER_BASELINES:
        return pd.DataFrame()
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.linear_model import LogisticRegression

    print_block("CLINICAL BIOMARKER BASELINES")
    prep = LeakSafePreprocessor(representation="continuous")
    X_train_cont = prep.fit_transform(X_train_raw, y_train).reset_index(drop=True)
    X_test_cont = prep.transform(X_test_raw).reset_index(drop=True)
    y_train = y_train.reset_index(drop=True) if hasattr(y_train, "reset_index") else pd.Series(y_train).reset_index(drop=True)
    y_test = y_test.reset_index(drop=True) if hasattr(y_test, "reset_index") else pd.Series(y_test).reset_index(drop=True)

    feature_sets = {
        "CA125 alone": ["CA125"],
        "HE4 alone": ["HE4"],
        "CA125 + HE4": ["CA125", "HE4"],
        "CA125 + HE4 + Age": ["CA125", "HE4", "Age"],
        "CA125 + HE4 + Age + Menopause": ["CA125", "HE4", "Age", "Menopause"],
    }
    rows = []
    ci_tables = []
    for name, feats in feature_sets.items():
        available = [f for f in feats if f in X_train_cont.columns]
        missing = [f for f in feats if f not in X_train_cont.columns]
        if missing or len(available) == 0:
            rows.append({"Baseline": name, "Missing_Features": ";".join(missing), "Error": "Required features unavailable after preprocessing."})
            continue
        model = Pipeline([
            ("scaler", StandardScaler()),
            ("lr", LogisticRegression(max_iter=5000, solver="lbfgs", random_state=RANDOM_STATE)),
        ])
        try:
            model.fit(X_train_cont[available], y_train)
            p = proba_positive(model, X_test_cont[available])
            pred = (p >= 0.5).astype(int)
            row = {
                "Baseline": name,
                "Model": "Standardized Logistic Regression",
                "Features": ";".join(available),
                "Num_Features": len(available),
                **metrics_from_predictions(y_test, pred, p),
                **calibration_intercept_slope(y_test, p),
            }
            rows.append(row)
            metric_line(f"CLINICAL BIOMARKER BASELINE | {name}", row)
            ci = bootstrap_metric_ci(y_test, pred, p)
            ci.insert(0, "Baseline", name)
            ci_tables.append(ci)
        except Exception as e:
            rows.append({"Baseline": name, "Features": ";".join(available), "Error": str(e)})
            print(f"Clinical biomarker baseline failed | {name}: {e}")

    df = pd.DataFrame(rows)
    df.to_csv(out_dirs["tables"] / "clinical_biomarker_baselines.csv", index=False)
    print_df("Clinical biomarker baseline results", df, max_rows=None)
    if ci_tables:
        ci_df = pd.concat(ci_tables, ignore_index=True)
        ci_df.to_csv(out_dirs["tables"] / f"clinical_biomarker_baselines_bootstrap_{int(BOOTSTRAP_METRIC_CI_LEVEL*100)}ci.csv", index=False)
        print_df(f"Clinical biomarker baseline bootstrap {int(BOOTSTRAP_METRIC_CI_LEVEL*100)}% CI", ci_df, max_rows=None)
    return df


def final_train_test_evaluate_selector(
    selector_name: str,
    X_train_raw: Any,
    y_train: Any,
    X_test_raw: Any,
    y_test: Any,
    oof_store: Dict[str, Any],
    out_dirs: Dict[str, Path],
    representation: str = "binned",
) -> Tuple[Any, Dict[str, Any], Dict[str, Any]]:
    import pandas as pd
    import joblib

    print(f"\n===== Final train/test evaluation for {selector_name} =====")
    final_label = f"{selector_name}_final_train80"
    prep, selected_features, selector_info, X_train_prep, X_filtered, kw_features, cv_dropped, cv_mat = run_fold_preprocess_and_select(
        selector_name, X_train_raw.reset_index(drop=True), y_train.reset_index(drop=True), out_dirs, final_label, representation=representation
    )
    X_test_prep = prep.transform(X_test_raw.reset_index(drop=True)).reset_index(drop=True)
    print_selector_details(f"{selector_name} FINAL 80% TRAIN", selector_info)
    print(f"{selector_name} FINAL selected features ({len(selected_features)}): {selected_features}")
    print_preprocessing_details(
        f"{selector_name} FINAL 80% TRAIN",
        prep,
        X_train_raw.reset_index(drop=True),
        X_train_prep,
        kw_features,
        cv_dropped,
        X_filtered,
        selected_features,
    )

    # Save preprocessing audit fitted only on 80 percent train.
    prep.audit_dataframe().to_csv(out_dirs["tables"] / f"{selector_name}_preprocessing_parse_audit_train80.csv", index=False)
    with open(out_dirs["features"] / f"{selector_name}_feature_categories_train80.json", "w", encoding="utf-8") as f:
        json.dump(prep.feature_categories_, f, indent=2)

    # Plots for final selected features.
    if len(selected_features) > 1:
        plot_selected_correlation(
            X_train_prep[selected_features], y_train.reset_index(drop=True),
            f"{selector_name}: Correlation Heatmap of Final Selected Features and Target",
            out_dirs["figures"] / f"{selector_name}_final_selected_correlation_heatmap.png",
        )
        # Cramer's heatmap after Kruskal.
        if cv_mat is not None and cv_mat.shape[0] <= 35:
            plot_heatmap(
                cv_mat, f"{selector_name}: Cramer's V Heatmap After Kruskal Filter",
                out_dirs["figures"] / f"{selector_name}_final_cramers_v_heatmap.png",
            )

    if selector_name == "Stat-RS":
        plot_stat_rs_final_bubble_from_saved_rankings(out_dirs, final_label, selected_features)

    X_train_sel = X_train_prep[selected_features]
    X_test_sel = X_test_prep[selected_features]

    model_defs = model_spaces()
    test_rows = []
    ci_tables = []
    fitted_models: Dict[str, Any] = {}

    for model_name, spec in model_defs.items():
        print(f"Final tuning/evaluating {selector_name} | model {model_name}")
        try:
            search = tune_model(spec["estimator"], spec["params"], X_train_sel, y_train)
            best_model = search.best_estimator_
            fitted_models[model_name] = best_model
            p_test = proba_positive(best_model, X_test_sel)
            pred_test = (p_test >= 0.5).astype(int)
            met = metrics_from_predictions(y_test, pred_test, p_test)
            cal = calibration_intercept_slope(y_test, p_test)
            row = {
                "Selector": selector_name,
                "Representation": representation,
                "Model": model_name,
                "Num_Selected_Features": len(selected_features),
                "Selected_Features": ";".join(selected_features),
                "Best_Params": json.dumps(search.best_params_, default=str),
                **met,
                **cal,
            }
            test_rows.append(row)
            metric_line(f"HELD-OUT TEST RESULT | {selector_name} | model={model_name}", row)
            print(f"Best params | {selector_name} | {model_name}: {search.best_params_}")
            if PLOT_CALIBRATION_FOR_EACH_MODEL:
                plot_calibration(
                    y_test, p_test,
                    f"{selector_name} + {model_name}: Held-out Test Calibration",
                    out_dirs["figures"] / f"{selector_name}_{model_name}_heldout_calibration_all_models.png",
                )

            ci = bootstrap_metric_ci(y_test, pred_test, p_test)
            ci.insert(0, "Model", model_name)
            ci.insert(0, "Selector", selector_name)
            ci_tables.append(ci)

            joblib.dump(best_model, out_dirs["models"] / f"{selector_name}_{model_name}_best_model.pkl")
        except Exception as e:
            print(f"Final model failed: {selector_name} model {model_name}: {e}")
            test_rows.append({"Selector": selector_name, "Representation": representation, "Model": model_name, "Error": str(e)})

    test_df = pd.DataFrame(test_rows)
    test_df.to_csv(out_dirs["tables"] / f"{selector_name}_heldout_test_metrics.csv", index=False)
    print_df(f"{selector_name}: HELD-OUT TEST RESULTS FOR ALL MODELS", test_df, max_rows=None)
    if ci_tables:
        ci_df = pd.concat(ci_tables, ignore_index=True)
        ci_df.to_csv(out_dirs["tables"] / f"{selector_name}_heldout_test_bootstrap_ci.csv", index=False)
        print_df(f"{selector_name}: HELD-OUT TEST BOOTSTRAP {int(BOOTSTRAP_METRIC_CI_LEVEL*100)}% CI FOR ALL MODELS", ci_df, max_rows=None)

    # Choose best model prospectively by primary endpoint: held-out ROC_AUC, tie-breaker Log_Loss lower, then F1.
    valid = test_df.dropna(subset=["ROC_AUC"]).copy()
    valid["Log_Loss_sort"] = pd.to_numeric(valid.get("Log_Loss"), errors="coerce")
    valid["F1_sort"] = pd.to_numeric(valid.get("F1"), errors="coerce")
    valid = valid.sort_values(["ROC_AUC", "Log_Loss_sort", "F1_sort"], ascending=[False, True, False])
    best_info: Dict[str, Any] = {}
    if not valid.empty:
        best_row = valid.iloc[0].to_dict()
        best_model_name = best_row["Model"]
        best_model = fitted_models.get(best_model_name)
        if best_model is not None:
            p_test = proba_positive(best_model, X_test_sel)
            pred_test = (p_test >= 0.5).astype(int)
            if best_model_name in oof_store and len(oof_store[best_model_name]["y_true"]) > 0:
                plot_roc_comparison(
                    oof_store[best_model_name]["y_true"], oof_store[best_model_name]["y_prob"],
                    y_test, p_test,
                    f"{selector_name} + {best_model_name}: Base Model Nested CV vs Held-out Test ROC",
                    out_dirs["figures"] / f"{selector_name}_{best_model_name}_BASE_MODEL_roc_nested_vs_test.png",
                )
                plot_confusion_matrices(
                    oof_store[best_model_name]["y_true"], oof_store[best_model_name]["y_pred"],
                    y_test, pred_test,
                    f"{selector_name} + {best_model_name}: Base Model Confusion Matrices",
                    out_dirs["figures"] / f"{selector_name}_{best_model_name}_BASE_MODEL_confusion_matrices.png",
                )
            best_cal = plot_calibration(
                y_test, p_test,
                f"{selector_name} + {best_model_name}: Base Model Held-out Test Calibration",
                out_dirs["figures"] / f"{selector_name}_{best_model_name}_BASE_MODEL_heldout_calibration.png",
            )
            plot_predicted_probability_distribution(
                y_test, p_test,
                f"{selector_name} + {best_model_name}: Predicted Probability Distribution",
                out_dirs["figures"] / f"{selector_name}_{best_model_name}_BASE_MODEL_predicted_probability_distribution.png",
            )

            # Separate ten-fold nested-CV table for the selected/base model only.
            nested_fold_path = out_dirs["tables"] / f"{selector_name}_nested_cv_fold_metrics.csv"
            if nested_fold_path.exists():
                nested_fold_df = pd.read_csv(nested_fold_path)
                base_fold_df = nested_fold_df[nested_fold_df["Model"] == best_model_name].copy()
                base_fold_df.to_csv(out_dirs["tables"] / f"{selector_name}_{best_model_name}_BASE_MODEL_nested_cv_10fold_results.csv", index=False)
                keep_cols = [c for c in [
                    "Selector", "Fold", "Model", "Num_Selected_Features",
                    "Accuracy", "Sensitivity_Recall", "Specificity", "PPV_Precision", "NPV",
                    "F1", "ROC_AUC", "Log_Loss", "Brier", "TN", "FP", "FN", "TP",
                    "Selected_Features"
                ] if c in base_fold_df.columns]
                print_df(f"{selector_name} + {best_model_name}: BASE MODEL NESTED-CV 10-FOLD RESULTS ONLY", base_fold_df[keep_cols], max_rows=None)

            best_ci = bootstrap_metric_ci(y_test, pred_test, p_test)
            best_ci.insert(0, "Model", best_model_name)
            best_ci.insert(0, "Selector", selector_name)
            best_ci.to_csv(out_dirs["tables"] / f"{selector_name}_{best_model_name}_BASE_MODEL_heldout_bootstrap_{int(BOOTSTRAP_METRIC_CI_LEVEL*100)}ci.csv", index=False)
            print_df(f"{selector_name} + {best_model_name}: BASE MODEL HELD-OUT BOOTSTRAP {int(BOOTSTRAP_METRIC_CI_LEVEL*100)}% CI", best_ci, max_rows=None)

            best_performance_row = dict(best_row)
            best_performance_row.update({f"calibration_{k}": v for k, v in best_cal.items()})
            best_performance_df = pd.DataFrame([best_performance_row])
            best_performance_df.to_csv(out_dirs["tables"] / f"{selector_name}_{best_model_name}_BASE_MODEL_performance_calibration_summary.csv", index=False)
            print_df(f"{selector_name} + {best_model_name}: BASE MODEL PERFORMANCE + CALIBRATION SUMMARY", best_performance_df, max_rows=None)

            # LR baselines are not used as the main baseline now because the selected base model is usually XGB.
            # For the proposed Stat-AQ branch, run XGB variant baselines on the same selected feature space.
            if selector_name == "Stat-AQ":
                run_stat_aq_xgb_variant_baselines(
                    selector_name, X_train_sel, y_train, X_test_sel, y_test, out_dirs,
                    tuned_best_model=best_model if best_model_name == "XGB" else None,
                )
            if best_model_name == "LR" and RUN_LR_BASELINES_FOR_EACH_SELECTOR:
                print(f"{selector_name}: LR is the selected base/best model, so LR baseline comparison will be run on this selector's selected features.")
                run_lr_baseline_comparison(selector_name, X_train_sel, y_train, X_test_sel, y_test, out_dirs)

            explain_best_model_with_xai(
                selector_name, best_model_name, best_model,
                X_train_sel.reset_index(drop=True), X_test_sel.reset_index(drop=True),
                y_test.reset_index(drop=True) if hasattr(y_test, "reset_index") else y_test,
                selected_features, out_dirs,
            )
            best_info = {"best_model": best_model_name, "best_row": best_row, "calibration": best_cal}
            print_block(f"{selector_name}: SELECTED BEST MODEL")
            print(f"Best model: {best_model_name}")
            metric_line(f"BEST MODEL METRICS | {selector_name} | {best_model_name}", best_row)
            with open(out_dirs["tables"] / f"{selector_name}_best_model.json", "w", encoding="utf-8") as f:
                json.dump(best_info, f, indent=2, default=str)

    # Save selected features and selector info.
    pd.DataFrame({"selected_feature": selected_features}).to_csv(out_dirs["features"] / f"{selector_name}_FINAL_selected_features.csv", index=False)
    with open(out_dirs["features"] / f"{selector_name}_FINAL_selector_info.json", "w", encoding="utf-8") as f:
        json.dump(selector_info, f, indent=2, default=str)

    return test_df, best_info, selector_info


def sensitivity_analysis(
    X_train_raw: Any,
    y_train: Any,
    X_test_raw: Any,
    y_test: Any,
    out_dirs: Dict[str, Path],
) -> Any:
    """
    Required representation sensitivity analysis:
    - current binned representation;
    - continuous standardized LR;
    - spline-expanded LR.

    Clinical cutoffs are included only if CLINICAL_CUTOFFS is populated with validated cutoffs.
    """
    import pandas as pd
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler, SplineTransformer
    from sklearn.linear_model import LogisticRegression

    rows = []
    representations = ["binned", "continuous"]
    if CLINICAL_CUTOFFS:
        representations.append("clinical_cutoff")

    for rep in representations:
        print(f"\nSensitivity analysis representation: {rep}")
        prep = LeakSafePreprocessor(representation=rep)
        Xtr = prep.fit_transform(X_train_raw, y_train).reset_index(drop=True)
        Xte = prep.transform(X_test_raw).reset_index(drop=True)
        ytr = y_train.reset_index(drop=True)
        yte = y_test.reset_index(drop=True)

        kw_features, _ = kruskal_filter(Xtr, ytr)
        kept, dropped, _ = cramers_v_filter(Xtr[kw_features])
        if not kept:
            kept = list(Xtr.columns)
        Xtr2 = Xtr[kept]
        Xte2 = Xte[kept]

        lr = Pipeline([
            ("scaler", StandardScaler()),
            ("lr", LogisticRegression(max_iter=2000, solver="lbfgs", random_state=RANDOM_STATE)),
        ])
        lr.fit(Xtr2, ytr)
        p = proba_positive(lr, Xte2)
        pred = (p >= 0.5).astype(int)
        rows.append({"Representation": rep, "Model": "LR_standardized", "Num_Features": len(kept), **metrics_from_predictions(yte, pred, p)})

        if rep == "continuous":
            spline_lr = Pipeline([
                ("scaler", StandardScaler()),
                ("spline", SplineTransformer(n_knots=4, degree=3, include_bias=False)),
                ("lr", LogisticRegression(max_iter=3000, solver="lbfgs", random_state=RANDOM_STATE)),
            ])
            try:
                spline_lr.fit(Xtr2, ytr)
                p2 = proba_positive(spline_lr, Xte2)
                pred2 = (p2 >= 0.5).astype(int)
                rows.append({"Representation": "continuous_spline", "Model": "Spline_LR", "Num_Features": len(kept), **metrics_from_predictions(yte, pred2, p2)})
            except Exception as e:
                rows.append({"Representation": "continuous_spline", "Model": "Spline_LR", "Error": str(e)})

    sens_df = pd.DataFrame(rows)
    sens_df.to_csv(out_dirs["tables"] / "representation_sensitivity_analysis.csv", index=False)
    print("\nRepresentation sensitivity analysis:")
    print(sens_df)
    return sens_df


def combine_final_summary(out_dirs: Dict[str, Path]) -> None:
    import pandas as pd

    tables = []
    for selector in ["Stat-RS", "Stat-AQ"]:
        test_path = out_dirs["tables"] / f"{selector}_heldout_test_metrics.csv"
        cv_path = out_dirs["tables"] / f"{selector}_nested_cv_summary.csv"
        if test_path.exists():
            df = pd.read_csv(test_path)
            df["Evaluation"] = "Heldout_Test"
            tables.append(df)
        if cv_path.exists():
            df = pd.read_csv(cv_path)
            df["Evaluation"] = "Nested_CV_Summary"
            tables.append(df)
    if tables:
        final = pd.concat(tables, ignore_index=True, sort=False)
        final.to_csv(out_dirs["tables"] / "ALL_MODEL_RESULTS_COMBINED.csv", index=False)
        print("\nFinal combined model results saved to:", out_dirs["tables"] / "ALL_MODEL_RESULTS_COMBINED.csv")
        print_df("COMBINED RESULTS: BOTH SELECTORS, ALL MODELS", final, max_rows=None)


def main() -> None:
    set_seed(RANDOM_STATE)
    ensure_packages()
    out_dirs = make_output_dirs(OUTPUT_DIR)

    df_raw, source_path = load_raw_dataset()
    print_and_save_dataset_report(df_raw, out_dirs["root"], source_path)
    save_enhanced_dataset_diagnostics(df_raw, out_dirs)

    # Split BEFORE fitting imputation, binning, Kruskal, Cramer's V, RFE, SHAP, ANN, QUBO, tuning, or fitting.
    X_train_raw, X_test_raw, y_train, y_test = split_raw_holdout(df_raw)
    X_train_raw, y_train = reset_xy(X_train_raw, y_train)
    X_test_raw, y_test = reset_xy(X_test_raw, y_test)

    split_report = {
        "raw_total_rows": int(len(df_raw)),
        "train_rows": int(len(X_train_raw)),
        "heldout_test_rows": int(len(X_test_raw)),
        "train_class_counts": y_train.value_counts().sort_index().to_dict(),
        "heldout_test_class_counts": y_test.value_counts().sort_index().to_dict(),
        "leakage_guard": "All fitted preprocessing, statistical filtering, feature selection, tuning, and model fitting occur after this split and inside each outer-training fold during nested CV.",
    }
    print("\nSPLIT REPORT")
    print(json.dumps(split_report, indent=2))
    with open(out_dirs["root"] / "split_report.json", "w", encoding="utf-8") as f:
        json.dump(split_report, f, indent=2)

    all_best = {}
    for selector in ["Stat-RS", "Stat-AQ"]:
        fold_df, oof_store, stability = nested_cv_evaluate_selector(
            selector, X_train_raw, y_train, out_dirs, representation="binned"
        )
        test_df, best_info, selector_info = final_train_test_evaluate_selector(
            selector, X_train_raw, y_train, X_test_raw, y_test, oof_store, out_dirs, representation="binned"
        )
        all_best[selector] = best_info

    run_clinical_biomarker_baselines(X_train_raw, y_train, X_test_raw, y_test, out_dirs)
    sensitivity_analysis(X_train_raw, y_train, X_test_raw, y_test, out_dirs)
    combine_final_summary(out_dirs)

    with open(out_dirs["tables"] / "BEST_MODELS_BY_SELECTOR.json", "w", encoding="utf-8") as f:
        json.dump(all_best, f, indent=2, default=str)

    print_df("BEST MODELS BY SELECTOR", pd.DataFrame([{"Selector": k, "Best_Model": v.get("best_model"), **{kk: vv for kk, vv in v.get("best_row", {}).items() if kk in ["Accuracy", "Sensitivity_Recall", "Specificity", "PPV_Precision", "NPV", "F1", "ROC_AUC", "Log_Loss", "Brier", "TN", "FP", "FN", "TP"]}} for k, v in all_best.items()]), max_rows=None)
    print_output_inventory(out_dirs)
    if DISPLAY_ALL_SAVED_IMAGES_AT_END:
        display_all_saved_images(out_dirs)

    print("\nDONE.")
    print(f"All outputs are in: {Path(OUTPUT_DIR).resolve()}")
    print("Key output tables/figures were also saved internally; images are displayed in the notebook output cell as they are created.")
    print("Key files:")
    print("  tables/ALL_MODEL_RESULTS_COMBINED.csv")
    print("  tables/representation_sensitivity_analysis.csv")
    print("  features/Stat-RS_FINAL_selected_features.csv")
    print("  features/Stat-AQ_FINAL_selected_features.csv")
    print("  tables/Stat-RS_heldout_test_bootstrap_ci.csv")
    print("  tables/Stat-AQ_heldout_test_bootstrap_ci.csv")
    print("  tables/*BASE_MODEL_heldout_bootstrap_*ci.csv")
    print("  tables/*BASE_MODEL_performance_calibration_summary.csv")
    print("  figures/*FINAL_stat_rs_intersection_bubble.png")
    print("  figures/*stat_aq_final_union_radar.png")
    print("  figures/*BASE_MODEL_roc_nested_vs_test.png")
    print("  figures/*BASE_MODEL_confusion_matrices.png")
    print("  figures/*BASE_MODEL_heldout_calibration.png")
    print("  figures/*BASE_MODEL_predicted_probability_distribution.png")


if __name__ == "__main__":
    main()
```
